# Data Cleaning & Modeling Drills

Each drill hands you **messy raw data** in a sandbox. Your job is to write SQL that
**cleans** it and **builds the right model** (`CREATE TABLE ... AS SELECT ...`).

You're graded differently from the query notebook: instead of one result set, you
**build tables**, and a **check query** is run against *your* tables and compared to the
expected answer. So you're graded on the model you built, not a one-off query.

**How to use it**
1. Run the setup cell.
2. Pick a drill, run the *show* cell to see the messy data, the task, and the principles.
3. Write your `CREATE TABLE` statements in the *your solution* cell.
4. Run the *grade* cell. Fix until it says ✅.
5. Stuck? reveal the reference solution at the bottom.
