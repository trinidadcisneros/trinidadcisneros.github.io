# Phase 1: Data Acquisition

## Maternal Care Provider Network Analysis

---

### Project Overview

This notebook handles the acquisition, validation, and initial profiling of all datasets required for our OB-GYN provider network analysis. The goal is to map maternal care provider referral networks across the U.S., identify regions with strong vs. fragmented care coordination, and overlay maternal health outcomes to reveal where virtual care models could fill the biggest gaps.

### Datasets

| # | Dataset | Source | Study Period | Description |
|---|---------|--------|-------------|-------------|
| 1 | CMS Physician Shared Patient Patterns | NBER | 2015 (60-day window) | Edge list of provider pairs sharing patients |
| 2 | NPPES NPI Bulk File | CMS | Current (monthly) | Provider registry with specialty and location |
| 3 | NUCC Taxonomy Code Set | NUCC | Current (v25.1) | Crosswalk of taxonomy codes to specialty names |
| 4 | CDC WONDER Natality Expanded | CDC | 2016 | County-level maternal morbidity indicators |

### Study Period Rationale

The CMS Shared Patient Patterns data (2009–2015) is the binding constraint. We use **2015** as the most recent and complete year. The **60-day window** captures referral-to-completion lag time — the realistic gap between when an OB-GYN refers a patient and when the specialist visit actually occurs. CDC WONDER data starts at **2016**, which aligns well since CDC natality statistics for a given year reflect births (and associated morbidity) from the prior reporting period.

### Assumptions

- **NPPES Current vs. Historical:** A 2015-era NPPES snapshot is not freely available from CMS or NBER. We use the current NPPES bulk file under the assumption that OB-GYN taxonomy codes (207V) have not changed and that the majority of established practices have stable locations. This is noted as a limitation.
- **Privacy Threshold:** The CMS shared patient file excludes provider pairs with fewer than 11 shared patients, meaning very low-volume referral relationships are not captured.

## 2. Dataset 1 — CMS Physician Shared Patient Patterns (2015, 60-Day Window)

**Source:** [NBER Physician Shared Patient Patterns Data](https://www.nber.org/research/data/physician-shared-patient-patterns-data)

**File:** `physician-shared-patient-patterns-2015-days60.txt`

**Format:** Headerless, comma-delimited text file with 5 columns:

| Column Position | Field | Description |
|----------------|-------|-------------|
| 1 | `npi_1` | NPI of the initial/referring provider |
| 2 | `npi_2` | NPI of the subsequent/referred-to provider |
| 3 | `shared_count` | Total shared patient encounters in the 60-day rolling window |
| 4 | `same_day_count` | Number of same-day shared encounters |
| 5 | `unique_bene_count` | Number of unique beneficiaries (patients) shared |

**Note:** Pairs with fewer than 11 shared patients are excluded per CMS privacy rules.

**Expected size:** ~2–3 GB

```python
# -------------------------------------------------------------------
# DOWNLOAD: CMS Shared Patient Patterns 2015 (60-day window)
# -------------------------------------------------------------------
# The NBER hosts these files at:
# https://data.nber.org/physician-shared-patient-patterns/
#
# If the direct download below does not work (NBER may require
# browser-based access), download manually from:
# https://www.nber.org/research/data/physician-shared-patient-patterns-data
#
# Navigate to the 2015 data and select the 60-day file.
# Save it to: {CMS_DIR}/physician-shared-patient-patterns-2015-days60.txt
# -------------------------------------------------------------------

CMS_FILENAME = "physician-shared-patient-patterns-2015-days60.txt"
CMS_FILEPATH = CMS_DIR / CMS_FILENAME

# Attempt programmatic download
CMS_URL = "https://data.nber.org/physician-shared-patient-patterns/2015/physician-shared-patient-patterns-2015-days60.txt"

if CMS_FILEPATH.exists():
    print(f"File already exists: {CMS_FILEPATH}")
    print(f"File size: {CMS_FILEPATH.stat().st_size / (1024**3):.2f} GB")
else:
    print(f"Attempting download from: {CMS_URL}")
    print("NOTE: This is a large file (~2.2 GB). Download may take 10-20 minutes.")
    print("If this fails, download manually from the NBER link above.\n")

    MAX_RETRIES = 3
    CHUNK_SIZE = 1024 * 1024  # 1 MB chunks (smaller = more resilient to drops)
    CONNECT_TIMEOUT = 60      # seconds to establish connection
    READ_TIMEOUT = 300        # seconds to wait between chunks (5 min)

    for attempt in range(1, MAX_RETRIES + 1):
        try:
            print(f"--- Attempt {attempt} of {MAX_RETRIES} ---")
            response = requests.get(
                CMS_URL,
                stream=True,
                timeout=(CONNECT_TIMEOUT, READ_TIMEOUT),
                headers={"User-Agent": "Mozilla/5.0 (research-download)"}
            )
            response.raise_for_status()
            total_size = int(response.headers.get('content-length', 0))
            downloaded = 0

            with open(CMS_FILEPATH, 'wb') as f:
                for chunk in response.iter_content(chunk_size=CHUNK_SIZE):
                    f.write(chunk)
                    downloaded += len(chunk)
                    if total_size > 0:
                        pct = (downloaded / total_size) * 100
                        print(f"\rProgress: {pct:.1f}% ({downloaded/(1024**3):.2f} / {total_size/(1024**3):.2f} GB)", end="")

            print(f"\n\nDownload complete: {CMS_FILEPATH}")
            print(f"File size: {CMS_FILEPATH.stat().st_size / (1024**3):.2f} GB")
            break  # Success — exit retry loop

        except Exception as e:
            print(f"\n\nAttempt {attempt} failed: {e}")

            # Clean up partial file
            if CMS_FILEPATH.exists():
                CMS_FILEPATH.unlink()
                print("Partial file removed.")

            if attempt < MAX_RETRIES:
                import time
                wait = 30 * attempt  # Back off: 30s, 60s, 90s
                print(f"Waiting {wait}s before retry...\n")
                time.sleep(wait)
            else:
                print(f"\n*** ALL {MAX_RETRIES} ATTEMPTS FAILED — MANUAL DOWNLOAD REQUIRED ***")
                print(f"1. Go to: https://www.nber.org/research/data/physician-shared-patient-patterns-data")
                print(f"2. Download the 2015, 60-day window file")
                print(f"3. Save to: {CMS_FILEPATH}")
```

```python
# -------------------------------------------------------------------
# VALIDATE: Read first rows and confirm structure
# -------------------------------------------------------------------

CMS_COLUMNS = ["npi_1", "npi_2", "shared_count", "same_day_count", "unique_bene_count"]

if CMS_FILEPATH.exists():
    # Read only first 10 rows to validate structure
    cms_sample = pd.read_csv(
        CMS_FILEPATH,
        header=None,
        names=CMS_COLUMNS,
        nrows=10,
        dtype={"npi_1": str, "npi_2": str}
    )
    print("CMS Shared Patient Patterns — First 10 rows:")
    print(f"Columns: {list(cms_sample.columns)}")
    print(f"Shape: {cms_sample.shape}")
    display(cms_sample)
    
    # Confirm data types
    print("\nData types:")
    print(cms_sample.dtypes)
else:
    print(f"File not found: {CMS_FILEPATH}")
    print("Please complete the manual download step above.")
```

```python
# -------------------------------------------------------------------
# PROFILE: Get total row count (without loading full file into memory)
# -------------------------------------------------------------------

if CMS_FILEPATH.exists():
    print("Counting total rows (this may take a minute for a multi-GB file)...")
    row_count = 0
    with open(CMS_FILEPATH, 'r') as f:
        for _ in f:
            row_count += 1
    print(f"Total rows: {row_count:,}")
    print(f"File size: {CMS_FILEPATH.stat().st_size / (1024**3):.2f} GB")
else:
    print("File not yet downloaded.")
```

## 3. Dataset 2 — NPPES NPI Bulk File

**Source:** [CMS NPPES Downloads](https://download.cms.gov/nppes/NPI_Files.html)

**File:** Full Replacement Monthly NPI File (ZIP → CSV)

**Key columns we will use:**

| Column Name | Description |
|------------|-------------|
| `NPI` | National Provider Identifier (join key to CMS shared patient data) |
| `Entity Type Code` | 1 = Individual, 2 = Organization |
| `Provider Last Name (Legal Name)` | Provider surname |
| `Provider First Name` | Provider first name |
| `Healthcare Provider Taxonomy Code_1` through `_15` | Up to 15 taxonomy codes per provider |
| `Provider Business Practice Location Address City Name` | Practice city |
| `Provider Business Practice Location Address State Name` | Practice state |
| `Provider Business Practice Location Address Postal Code` | Practice ZIP code |

**Assumption:** We use the current NPPES file because CMS does not archive historical snapshots and the NBER archive for 2015 is incomplete. OB-GYN taxonomy codes (207V) have not changed, and established practice locations are generally stable. This is documented as a study limitation.

**Expected size:** ~1 GB compressed / ~9.5 GB uncompressed

```python
# -------------------------------------------------------------------
# DOWNLOAD: NPPES NPI Full Replacement Monthly File
# -------------------------------------------------------------------
# The file is available at:
# https://download.cms.gov/nppes/NPI_Files.html
#
# The download page provides a link to the latest full replacement
# ZIP file. The filename changes monthly (e.g., NPPES_Data_Dissemination_February_2026.zip).
#
# Due to file size (~1 GB) and CMS server requirements, 
# MANUAL DOWNLOAD IS RECOMMENDED:
#
# 1. Go to: https://download.cms.gov/nppes/NPI_Files.html
# 2. Click the "Full Replacement Monthly NPI File" download link
# 3. Save the ZIP file to the NPPES_DIR path shown below
# 4. The next cell will handle extraction
# -------------------------------------------------------------------

print(f"Save the NPPES ZIP file to: {NPPES_DIR}/")
print(f"\nDownload from: https://download.cms.gov/nppes/NPI_Files.html")
print(f"Look for: 'Full Replacement Monthly NPI File'")
```

```python
# -------------------------------------------------------------------
# EXTRACT: Unzip the NPPES file
# -------------------------------------------------------------------

# Find the zip file in the NPPES directory
nppes_zips = list(NPPES_DIR.glob("*.zip"))

if nppes_zips:
    nppes_zip = nppes_zips[0]  # Use the first zip found
    print(f"Found ZIP file: {nppes_zip.name}")
    print(f"ZIP size: {nppes_zip.stat().st_size / (1024**3):.2f} GB")
    
    # List contents before extracting
    with zipfile.ZipFile(nppes_zip, 'r') as z:
        print(f"\nContents of ZIP:")
        for info in z.infolist():
            print(f"  {info.filename} — {info.file_size / (1024**3):.2f} GB")
    
    # Extract (this will take a few minutes)
    print(f"\nExtracting to: {NPPES_DIR}/")
    print("This may take several minutes due to file size (~9.5 GB uncompressed)...")
    with zipfile.ZipFile(nppes_zip, 'r') as z:
        z.extractall(NPPES_DIR)
    print("Extraction complete.")
else:
    print(f"No ZIP file found in {NPPES_DIR}/")
    print("Please download the NPPES file first (see cell above).")
```

```python
# -------------------------------------------------------------------
# VALIDATE: Read a sample and confirm the columns we need exist
# -------------------------------------------------------------------

# The key columns we need for this project
NPPES_COLS_NEEDED = [
    "NPI",
    "Entity Type Code",
    "Provider Last Name (Legal Name)",
    "Provider First Name",
    "Provider Business Practice Location Address City Name",
    "Provider Business Practice Location Address State Name",
    "Provider Business Practice Location Address Postal Code",
]
# Add all 15 taxonomy code columns
TAXONOMY_COLS = [f"Healthcare Provider Taxonomy Code_{i}" for i in range(1, 16)]
NPPES_COLS_NEEDED.extend(TAXONOMY_COLS)

# Find the main CSV file (the largest one, starts with 'npidata')
nppes_csvs = list(NPPES_DIR.glob("npidata*.csv"))

if nppes_csvs:
    nppes_csv = nppes_csvs[0]
    print(f"Found NPPES CSV: {nppes_csv.name}")
    print(f"File size: {nppes_csv.stat().st_size / (1024**3):.2f} GB")
    
    # Read only first 5 rows to validate column names
    nppes_sample = pd.read_csv(nppes_csv, nrows=5, dtype=str, low_memory=False)
    
    print(f"\nTotal columns in file: {len(nppes_sample.columns)}")
    print(f"\nChecking required columns exist:")
    missing = []
    for col in NPPES_COLS_NEEDED:
        status = "FOUND" if col in nppes_sample.columns else "MISSING"
        if status == "MISSING":
            missing.append(col)
        print(f"  {status}: {col}")
    
    if missing:
        print(f"\nWARNING: {len(missing)} required columns not found.")
        print("Column names may differ slightly. Here are all columns:")
        for c in nppes_sample.columns:
            print(f"  {c}")
    else:
        print(f"\nAll {len(NPPES_COLS_NEEDED)} required columns found.")
    
    # Show sample of our key columns
    display(nppes_sample[NPPES_COLS_NEEDED[:7]].head())
else:
    print(f"No NPPES CSV found in {NPPES_DIR}/")
    print("Please extract the ZIP file first (see cell above).")
```

## 4. Dataset 3 — NUCC Taxonomy Code Set

**Source:** [NUCC Provider Taxonomy CSV](https://www.nucc.org/index.php/code-sets-mainmenu-41/provider-taxonomy-mainmenu-40/csv-mainmenu-57)

**Purpose:** This crosswalk maps taxonomy codes to human-readable specialty names. We use it to identify all taxonomy codes under the **207V** family (Obstetrics & Gynecology), which includes:

| Taxonomy Code | Specialty |
|--------------|----------|
| `207V00000X` | Obstetrics & Gynecology (General) |
| `207VB0002X` | Bariatric Medicine |
| `207VC0200X` | Critical Care Medicine |
| `207VE0102X` | Reproductive Endocrinology |
| `207VF0040X` | Female Pelvic Medicine & Reconstructive Surgery |
| `207VG0400X` | Gynecology |
| `207VH0002X` | Hospice & Palliative Medicine |
| `207VM0101X` | Maternal & Fetal Medicine |
| `207VX0000X` | Obstetrics |
| `207VX0201X` | Gynecologic Oncology |

**Expected size:** < 1 MB

**Download:** Manual download required (NUCC requires license agreement acceptance via browser).

```python
# -------------------------------------------------------------------
# DOWNLOAD: NUCC Taxonomy Code Set (CSV)
# -------------------------------------------------------------------
# Manual download required — NUCC requires accepting a license
# agreement in the browser before downloading.
#
# 1. Go to: https://www.nucc.org/index.php/code-sets-mainmenu-41/provider-taxonomy-mainmenu-40/csv-mainmenu-57
# 2. Accept the license agreement
# 3. Download the CSV file
# 4. Save to the path shown below
# -------------------------------------------------------------------

NUCC_FILENAME = "nucc_taxonomy_251.csv"  # Rename to match version
NUCC_FILEPATH = NUCC_DIR / NUCC_FILENAME

print(f"Save the NUCC taxonomy CSV to: {NUCC_FILEPATH}")
print(f"\nDownload from: https://www.nucc.org/index.php/code-sets-mainmenu-41/provider-taxonomy-mainmenu-40/csv-mainmenu-57")
```

```python
# -------------------------------------------------------------------
# VALIDATE: Load taxonomy file and filter to OB-GYN codes (207V)
# -------------------------------------------------------------------

if NUCC_FILEPATH.exists():
    nucc_df = pd.read_csv(NUCC_FILEPATH, dtype=str)
    
    print(f"NUCC Taxonomy file loaded.")
    print(f"Total taxonomy codes: {len(nucc_df):,}")
    print(f"Columns: {list(nucc_df.columns)}")
    
    # Filter to OB-GYN codes (207V family)
    # The taxonomy code column name may vary — check and adapt
    code_col = [c for c in nucc_df.columns if 'code' in c.lower()][0]
    obgyn_codes = nucc_df[nucc_df[code_col].str.startswith("207V", na=False)]
    
    print(f"\nOB-GYN taxonomy codes found (207V family): {len(obgyn_codes)}")
    display(obgyn_codes)
else:
    print(f"File not found: {NUCC_FILEPATH}")
    print("Please complete the manual download step above.")
    print("\nUsing hardcoded OB-GYN codes as fallback:")
    OBGYN_TAXONOMY_CODES = [
        "207V00000X",  # Obstetrics & Gynecology
        "207VB0002X",  # Bariatric Medicine
        "207VC0200X",  # Critical Care Medicine
        "207VE0102X",  # Reproductive Endocrinology
        "207VF0040X",  # Female Pelvic Medicine & Reconstructive Surgery
        "207VG0400X",  # Gynecology
        "207VH0002X",  # Hospice & Palliative Medicine
        "207VM0101X",  # Maternal & Fetal Medicine
        "207VX0000X",  # Obstetrics
        "207VX0201X",  # Gynecologic Oncology
    ]
    for code in OBGYN_TAXONOMY_CODES:
        print(f"  {code}")
```

## 5. Dataset 4 — CDC WONDER Natality Expanded (2016)

**Source:** [CDC WONDER Natality Expanded 2016–2024](https://wonder.cdc.gov/natality-expanded-current.html)

**Purpose:** County-level maternal morbidity indicators to overlay with our network analysis. We use **2016** to align with the 2015 network data study period.

**Maternal morbidity indicators available:**
- Maternal Transfusion (Yes/No)
- Third or Fourth Degree Perineal Laceration (Yes/No)
- Ruptured Uterus (Yes/No)
- Unplanned Hysterectomy (Yes/No)
- Admission to Intensive Care Unit (Yes/No)

**Geographic granularity:** County-level (counties with population ≥ 100,000 per 2010 Census; smaller counties grouped as "Unidentified Counties" per state).

**Export limit:** CDC WONDER allows a maximum of 75,000 rows per query.

**Download:** This is a query-based export (not a direct file download). Follow the steps below.

```python
# -------------------------------------------------------------------
# DOWNLOAD: CDC WONDER Natality Expanded — Manual Query Export
# -------------------------------------------------------------------
# CDC WONDER requires an interactive query through the browser.
#
# STEP-BY-STEP INSTRUCTIONS:
#
# 1. Go to: https://wonder.cdc.gov/natality-expanded-current.html
# 2. Click "I Agree" to accept the data use agreement
#
# 3. Section 1 — Group Results By:
#    - Select: "County of Residence" (this gives state + county)
#    - Add: "Year" if you want multi-year data
#
# 4. Section 2 — Measures:
#    - Keep defaults (Births count)
#
# 5. Section 3 — Year:
#    - Select: "2016" only
#
# 6. Sections 4-10 — Leave defaults (All)
#
# 7. Section 11 — Maternal Morbidity:
#    - For EACH morbidity indicator, run a SEPARATE query:
#      a) Maternal Transfusion → select "Yes"
#      b) ICU Admission → select "Yes" 
#      c) (repeat for each indicator you want)
#    - OR: Select "At least one checked" under Maternal Morbidity
#      to get a combined morbidity flag
#
# 8. Click "Send" to run the query
#
# 9. On the results page:
#    - Click "Export Results" tab
#    - Save the .txt file to the CDC_DIR path below
#
# RECOMMENDED QUERY STRATEGY:
# Query 1: All births by county (2016) — baseline denominator
# Query 2: Births with "Maternal Morbidity = At least one checked" by county (2016)
# This gives you a morbidity rate per county.
# -------------------------------------------------------------------

CDC_FILENAME_BIRTHS = "cdc_wonder_2016_births_by_county.txt"
CDC_FILENAME_MORBIDITY = "cdc_wonder_2016_maternal_morbidity_by_county.txt"

CDC_BIRTHS_PATH = CDC_DIR / CDC_FILENAME_BIRTHS
CDC_MORBIDITY_PATH = CDC_DIR / CDC_FILENAME_MORBIDITY

print("CDC WONDER query-based export required.")
print(f"\nSave baseline births file to:     {CDC_BIRTHS_PATH}")
print(f"Save maternal morbidity file to:  {CDC_MORBIDITY_PATH}")
print(f"\nQuery URL: https://wonder.cdc.gov/natality-expanded-current.html")
```

```python
# -------------------------------------------------------------------
# VALIDATE: Load CDC WONDER exports
# -------------------------------------------------------------------

for label, fpath in [("Births", CDC_BIRTHS_PATH), ("Morbidity", CDC_MORBIDITY_PATH)]:
    if fpath.exists():
        # CDC WONDER exports are tab-delimited with metadata rows at the bottom
        # Read and stop at the "---" separator line
        lines = []
        with open(fpath, 'r') as f:
            for line in f:
                if line.startswith('"---'):
                    break
                lines.append(line)
        
        from io import StringIO
        cdc_df = pd.read_csv(StringIO(''.join(lines)), sep='\t', dtype=str)
        
        print(f"\n{label} file loaded: {fpath.name}")
        print(f"Rows: {len(cdc_df):,}")
        print(f"Columns: {list(cdc_df.columns)}")
        display(cdc_df.head())
    else:
        print(f"\n{label} file not found: {fpath}")
        print("Please complete the CDC WONDER query export above.")
```
