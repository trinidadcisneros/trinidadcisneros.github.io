**How to use it**
1. Run the cell below.
2. Pick a muscle group and press **Show stretches**.
3. Tick the stretches you want and press **Open selected**.
4. Each stretch opens with the video, photos (use the arrows to flip between them), and written steps.
5. Tick **I did this today** on the ones you complete, then press **Save today's session**.
6. **Show my history** shows what you did and what you missed, day by day.

**Short on time or decisions?** Use the **Full body routine** builder: pick how much time you have (5, 10, 15, or 30 minutes), press **Build my routine**, and work through the numbered list of drills it opens for you, top to bottom.
