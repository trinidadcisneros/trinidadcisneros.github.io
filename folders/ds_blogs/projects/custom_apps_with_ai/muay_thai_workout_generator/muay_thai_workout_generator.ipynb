# Muay Thai Workout Generator

Generates a ~40 min heavy-bag workout: fixed warmup + 3 burnouts + 7 drills that hit both your **difficulty mix** (basic / intermediate / advance) and a **category mix** (2 strikes, 2 lower body, 3 combinations). Output is **copy-ready for Google Keep**. Tick the drills you actually did to track them and favor **new** ones next time.

All logic lives in **`muay_thai_lib.py`**; this notebook just launches it.

### Notes

- **All logic is in `muay_thai_lib.py`** — edit it and re-run the load cell (auto-reloads).
- **Power use:** `mt.build_workout('advance')`, `mt.workout_text(...)`, `mt.show_workout(wk)`, `mt.mark_used('basic',[{...}])`, `mt.usage_counts()`, `mt.load_usage()`.
- **Tracking** writes `muay_thai_usage_log.csv` (drills/burnouts done) and `muay_thai_workout_log.csv` (each generated workout), next to the library.
- **Prefer new** excludes drills you've already marked done; if a valid mix isn't possible it falls back to the full set.
- A standalone, no-Python version also exists as `muay_thai_workout_generator.html`.
