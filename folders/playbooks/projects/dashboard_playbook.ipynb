# How to Build a Dashboard — From Request to Delivery

**A professional playbook for analysts who build dashboards in business settings.**

This guide covers the full lifecycle of a dashboard project — from the first stakeholder conversation to the final handoff. It is not a tool tutorial. It is a process framework: how to scope, design, build, validate, present, and document dashboards that earn trust, drive decisions, and survive after you move on.

**How to use this guide:**
- **Quick reference:** Jump to the Master Phase Table (Section 1) or Checklists (Section 9) for at-a-glance guidance
- **Decision support:** Use the Decision Tree (Section 2) to classify the request and pick the right approach
- **Deep learning:** Read the Phase Pattern Library (Section 3) for detailed guidance on each phase
- **Communication help:** Pull templates from Section 8 for stakeholder emails, scope confirmations, and walkthrough scripts
- **Mistake prevention:** Review Anti-Patterns (Section 6) before presenting or handing off

## 1. Master Phase Reference Table

Use this table as a quick-reference map. Every dashboard project moves through these phases — even if some are compressed for simple requests.

| Phase | Key Question | Primary Deliverable | Common Pitfall | Time Budget |
|-------|-------------|---------------------|----------------|-------------|
| **A. Scope** | What problem does this solve? | Written scope doc / confirmed Slack thread | Building before aligning on purpose | 15% |
| **B. Metrics & Data** | What numbers answer the question? | Metric definitions + source mapping | Using the wrong source of truth | 15% |
| **C. Design** | How should this look and flow? | Wireframe or sketch (paper/digital) | Jumping into the tool without a layout plan | 10% |
| **D. Build** | How do I construct this cleanly? | Working dashboard (draft) | No naming conventions, spaghetti calculated fields | 25% |
| **E. QA** | Can I trust every number on this? | Validated dashboard + QA log | Skipping validation because "it looks right" | 15% |
| **F. Present** | Does the stakeholder understand and trust it? | Walkthrough meeting + feedback notes | Showing the dashboard without telling the story | 5% |
| **G. Iterate** | What needs to change? | Revised dashboard | Accepting every request without scoping impact | 10% |
| **H. Handoff** | Can someone else maintain this? | Documentation + access provisioning | No documentation — dashboard becomes tribal knowledge | 5% |

> **The 15% rule:** If you spend less than 15% of your time on Scoping (A) and Metrics (B) combined, you are almost certainly building the wrong thing.

## 2. Decision Tree: What Kind of Dashboard Is This?

Before you design anything, classify the request. The type of dashboard determines scope, refresh cadence, design complexity, and stakeholder expectations.

```
┌─────────────────────────────────────────────────────┐
│          WHAT KIND OF DASHBOARD IS THIS?             │
└─────────────────┬───────────────────────────────────┘
                  │
                  ▼
┌─────────────────────────────────────────────────────┐
│  Will this be used repeatedly (weekly+)?            │
│                                                     │
│    YES ──────────────┐    NO ───────────────┐       │
└──────────────────────┘                      │       │
          │                                   ▼       │
          ▼                    ┌──────────────────┐   │
┌───────────────────────┐     │   AD-HOC VIEW     │   │
│  Who is the primary   │     │                    │   │
│  audience?            │     │ One-time answer    │   │
│                       │     │ Spreadsheet or     │   │
│  EXECUTIVES ──┐       │     │ quick viz is fine  │   │
│  MANAGERS ────┤       │     │ Minimal polish     │   │
│  OPERATIONS ──┘       │     │ Lifespan: days     │   │
└───────────────────────┘     └──────────────────┘
          │
    ┌─────┴──────────────────────┐
    │                            │
    ▼                            ▼
┌──────────────────┐  ┌───────────────────────┐
│  STRATEGIC       │  │  OPERATIONAL          │
│                  │  │                        │
│  Monthly/quarterly│ │  Daily or real-time    │
│  KPIs, trends,   │  │  Workflow monitoring   │
│  goal tracking   │  │  Alerts, thresholds    │
│  Exec audience   │  │  Team/manager audience │
│  High polish     │  │  High reliability      │
│  Fewer metrics   │  │  Many metrics          │
│  Lifespan: years │  │  Lifespan: months-years│
└──────────────────┘  └───────────────────────┘
```

### How the type affects your approach

| Dimension | Ad-Hoc | Operational | Strategic |
|-----------|--------|-------------|-----------|
| **Refresh** | None (snapshot) | Daily or real-time | Weekly / Monthly |
| **Audience** | Requester only | Team / managers | Directors / execs |
| **# of metrics** | 1–3 | 10–30 | 3–8 |
| **Design effort** | Low | Medium | High |
| **Documentation** | Optional | Required | Required |
| **QA rigor** | Spot check | Full validation | Full validation + exec review |
| **Iteration cycles** | 0–1 | 1–3 | 2–5 |
| **Tool** | Spreadsheet / quick SQL | BI tool | BI tool with polished design |

<a id='phase-a-scoping-the-request'></a>

### A. Scoping the Request

**Signals you're in this phase:** Someone says "can you build a dashboard for..." or "I need visibility into..." or "we need to track..."

**In plain language:** Before you touch any data or tool, you need to understand *why* this dashboard exists, *who* will use it, and *what decisions* it should inform. Most dashboard failures trace back to skipping or rushing this step.

---

**Intake Questions Checklist:**

| # | Question | Why It Matters |
|---|----------|----------------|
| 1 | What problem are you trying to solve? | Separates "I want a dashboard" from the actual need |
| 2 | What decisions will this dashboard help you make? | Reveals the real output — often not what was initially requested |
| 3 | Who else will use this? | Determines access, complexity, and language level |
| 4 | How often do you need to see this? (Daily? Weekly? Once?) | Classifies the dashboard type (see Decision Tree) |
| 5 | What are you using today to answer this question? | Exposes existing workarounds and data sources |
| 6 | Is there a deadline or event driving this? | Scopes the timeline and level of polish |
| 7 | Can you show me an example of what "good" looks like? | Anchors expectations and avoids mismatched visions |

---

**Scope Confirmation Template (email / Slack):**

> **Subject: Dashboard Scope Confirmation — [Project Name]**
>
> Hi [Name],
>
> Following up on our conversation — here's what I'm planning to build so we're aligned before I start:
>
> **Purpose:** [One sentence — what decision this supports]
> **Audience:** [Who will use it and how often]
> **Key metrics:** [3–8 bullet points]
> **Data sources:** [Where the numbers come from]
> **Out of scope:** [What this dashboard will NOT cover]
> **Timeline:** [When you'll see a first draft]
>
> Does this match your expectations? Any changes before I start building?

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| Building before scoping | Stakeholder says "just a quick dashboard" | Always send a scope confirmation, even for "quick" ones |
| Scope via hallway conversation | Verbal agreement has no paper trail | Follow up every conversation with a written summary |
| Saying yes to everything | Want to be helpful | Explicitly list what's out of scope |
| Assuming you know the audience | You've worked with this team before | Always ask — priorities shift, new people join |

<a id='phase-b-identifying-metrics-and-data-sources'></a>

### B. Identifying Metrics & Data Sources

**Signals you're in this phase:** You have a confirmed scope and need to figure out exactly *what numbers* go on the dashboard and *where they come from.*

**In plain language:** This is where you translate business questions into specific, queryable metrics and verify that your data sources can actually produce them reliably. Getting this wrong means building a beautiful dashboard with wrong or misleading numbers.

---

**Metric Definition Worksheet:**

For each metric on the dashboard, document:

| Field | Example |
|-------|---------|
| **Metric name** | Monthly Recurring Revenue (MRR) |
| **Business definition** | Sum of all active subscription revenue, normalized to monthly |
| **Calculation** | SUM(subscription_amount) WHERE status = 'active' AND type != 'one_time' |
| **Source table(s)** | billing.subscriptions, billing.invoices |
| **Grain** | One row per subscription per month |
| **Filters / exclusions** | Exclude trial accounts, internal test accounts |
| **Known quirks** | Refunds appear with 1-day lag; cancellations effective end-of-billing-period |
| **Source of truth owner** | Finance team / @jane in #revenue-data |

---

**The Vanity vs. Actionable Metric Test:**

```
┌──────────────────────────────────────────────────────────┐
│          IS THIS METRIC WORTH PUTTING ON THE DASHBOARD?  │
└──────────────┬───────────────────────────────────────────┘
               │
               ▼
┌──────────────────────────────────────────────────┐
│  Can someone make a different DECISION            │
│  based on whether this number goes up or down?    │
│                                                    │
│  YES ───────────────┐    NO ──────────────────┐   │
└─────────────────────┘                          │  │
          │                                      ▼  │
          ▼                        ┌────────────────┐
┌──────────────────┐              │ VANITY METRIC   │
│  ACTIONABLE      │              │                  │
│  Keep it         │              │ Drop it or move  │
│                  │              │ to appendix      │
└──────────────────┘              └────────────────┘
```

**Common vanity metrics to challenge:** Total page views (without context), total users ever (vs. active users), raw count of support tickets (without resolution rate), revenue without costs.

---

**Data Source Validation Checklist:**

| Check | How | Red Flag |
|-------|-----|----------|
| Row count sanity | COUNT(*) by date — any unexpected gaps or spikes? | Missing days, sudden 10x increases |
| NULL prevalence | COUNT(*) vs COUNT(column) for key fields | > 5% NULLs in a metric column |
| Freshness | MAX(updated_at) or MAX(event_date) | Data more than 24h stale without explanation |
| Duplicates | COUNT(*) vs COUNT(DISTINCT primary_key) | Any difference means duplicates exist |
| Cross-reference | Compare a known number (e.g., last month's revenue from finance) | > 2% variance needs investigation |
| Filter alignment | Do your WHERE clauses match the business definition? | "Active" means different things in different tables |

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| Using the wrong source of truth | Multiple tables contain "revenue" | Always ask the data owner which table is canonical |
| Metric definition drift | You and the stakeholder mean different things by "active users" | Write definitions down and get sign-off |
| Ignoring NULLs | They don't show up visually | Run NULL checks before building any chart |
| Stale data without disclosure | ETL ran late, dashboard shows yesterday's numbers | Add a "last refreshed" timestamp to every dashboard |

<a id='phase-c-designing-the-layout'></a>

### C. Designing the Layout

**Signals you're in this phase:** You know your metrics and data sources. Now you need to decide how to arrange them visually before opening your BI tool.

**In plain language:** Design on paper (or whiteboard) first. The biggest time sink in dashboard building is rearranging charts in the tool because you started without a layout plan. Five minutes of sketching saves an hour of drag-and-drop.

---

**Layout Principles:**

**The Inverted Pyramid:** Put the most important information at the top. Executives read dashboards like newspapers — headline first, details if they're interested.

```
┌───────────────────────────────────────────────┐
│         SCORECARD ROW (3-5 KPIs)              │
│   Revenue: $2.1M   |  MoM: +8%  |  Churn: 3% │
├───────────────────────────────────────────────┤
│                                               │
│         PRIMARY TREND CHART                   │
│   (The one chart that answers the main        │
│    question — usually a time series)          │
│                                               │
├────────────────────┬──────────────────────────┤
│                    │                          │
│  BREAKDOWN VIEW    │  COMPARISON VIEW         │
│  (by segment,      │  (vs. target, vs. prior  │
│   region, product)│   period, vs. cohort)    │
│                    │                          │
├────────────────────┴──────────────────────────┤
│                                               │
│         DETAIL TABLE / DRILL-DOWN             │
│   (For users who want the raw numbers)        │
│                                               │
└───────────────────────────────────────────────┘
```

---

**The 5-Second Rule:** A stakeholder glancing at your dashboard for 5 seconds should be able to answer: "Are things going well or badly?" If they can't, your hierarchy is wrong.

**The 3-Click Rule:** Any specific answer should be reachable within 3 filter selections or drill-down clicks. More than that means the dashboard is too deep or needs restructuring.

---

**Wireframe Checklist:**

| Element | Question to Answer |
|---------|-------------------|
| Title bar | Does it include: dashboard name, date range, last refreshed timestamp? |
| Filter bar | What filters exist? Are defaults sensible? (Don't default to "all time") |
| Scorecard row | What are the 3–5 headline numbers? Do they have comparison context (vs. last period, vs. target)? |
| Primary chart | What is the ONE chart that answers the main question? |
| Supporting charts | Do they add context or just add noise? (If in doubt, cut it) |
| Detail table | Is there a way to see underlying data for users who want specifics? |
| Mobile / narrow | Will this be viewed on a laptop? Phone? Projected in a meeting room? |

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| Chart junk | Filling space feels productive | Every chart must answer a specific question — state it as the chart title |
| No comparison context | Showing "Revenue: $2.1M" without "vs. what?" | Always include vs. prior period, vs. target, or vs. benchmark |
| Rainbow color palettes | Default tool colors are loud | Use 2–3 colors max. Reserve red/green for directional meaning only |
| Designing in the tool | It's tempting to start clicking | Sketch on paper first — even 60 seconds of sketching helps |

<a id='phase-d-building-the-dashboard'></a>

### D. Building the Dashboard

**Signals you're in this phase:** You have a wireframe, confirmed metrics, and validated data sources. Time to open the tool and build.

**In plain language:** This is the construction phase. The goal is to build something clean, maintainable, and performant — not just something that looks right on your screen today.

---

**Build Hygiene Checklist:**

| Practice | Why | Example |
|----------|-----|---------|
| **Naming conventions** | Someone else (or future you) needs to understand the logic | `calc_mrr_monthly` not `Calculation 1` |
| **Consistent date handling** | Mixed date grains cause silent errors | Decide: calendar month vs. rolling 30 days. Document it. |
| **Centralized calculations** | Same metric defined differently in two charts = credibility killer | Use calculated fields / parameters at the data source level, not per-chart |
| **Filter architecture** | Filters that affect some charts but not others confuse users | Map which filters apply to which views. Make it explicit. |
| **Performance testing** | A dashboard that takes 30s to load won't be used | Test with production data volumes. Aggregate where possible. |
| **Version snapshots** | You will need to roll back at some point | Save versioned copies: `dashboard_v1_draft`, `dashboard_v2_after_feedback` |

---

**Calculated Field Hygiene:**

```
┌──────────────────────────────────────────────────────┐
│  GOOD: Centralized, documented, reusable              │
│                                                       │
│  Name:  calc_mrr_monthly                              │
│  Logic: SUM(IF [status] = 'active'                    │
│              AND [type] != 'one_time'                  │
│              THEN [subscription_amount]                │
│              ELSE 0 END)                               │
│  Used in: Revenue Scorecard, Trend Chart, MoM Calc   │
│  Notes:  Excludes trials. Matches Finance definition. │
├──────────────────────────────────────────────────────┤
│  BAD: Inline, undocumented, duplicated                │
│                                                       │
│  Chart 1: SUM([subscription_amount])                  │
│  Chart 2: SUM(IF [type] != 'trial'                    │
│               THEN [subscription_amount] END)          │
│  Chart 3: SUM([amount]) -- different column entirely   │
│                                                       │
│  ▶ Three charts, three definitions, zero trust         │
└──────────────────────────────────────────────────────┘
```

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| "Calculation 1, Calculation 2..." | Default names are fast to create | Rename every calculated field immediately |
| Hardcoded date filters | "WHERE date >= '2025-01-01'" works today | Use relative dates: LAST 12 MONTHS, YEAR TO DATE |
| No null handling | NULLs silently change aggregations | Wrap metrics in COALESCE/IFNULL or explicitly filter |
| Over-engineering on v1 | Trying to build the final version first pass | Build the 80% version. Ship it. Iterate with real feedback. |

<a id='phase-e-qa-and-validation'></a>

### E. QA & Validation

**Signals you're in this phase:** The dashboard is built and looks right. Now you need to prove it *is* right.

**In plain language:** "Looks right" is not good enough. Every number on the dashboard needs to be verified against a source of truth. The first time a stakeholder finds a wrong number, the dashboard loses all credibility — and so do you.

---

**The QA Ladder (validate in this order):**

| Step | Check | Method |
|------|-------|--------|
| 1 | **Row counts** | Total rows in dashboard vs. COUNT(*) in source query |
| 2 | **Totals match** | Dashboard grand total vs. SUM() in SQL against source table |
| 3 | **Breakdowns add up** | Sum of segments = grand total (no double-counting, no gaps) |
| 4 | **Filters work correctly** | Apply each filter — does the number change as expected? |
| 5 | **Time boundaries** | First date and last date in dashboard match expected range |
| 6 | **NULL check** | Are NULLs excluded, included, or handled explicitly? |
| 7 | **Cross-reference** | Compare at least one metric to a known external number (finance report, prior deliverable) |
| 8 | **Edge cases** | What happens with zero values? New categories? A segment with one record? |
| 9 | **The screenshot test** | Can someone glancing for 5 seconds tell if things are good or bad? |
| 10 | **The "explain this number" test** | Point to any number. Can you explain exactly how it was calculated? |

---

**QA Log Template:**

| Metric | Dashboard Value | Source Query Value | Match? | Notes |
|--------|----------------|-------------------|--------|-------|
| Total MRR (March) | $2,143,500 | $2,143,500 | Yes | Verified against billing.subscriptions |
| Active Users (March) | 14,832 | 14,891 | No — 59 off | Discrepancy: dashboard excludes internal accounts, source query doesn't. Added filter. |
| Churn Rate | 3.2% | 3.2% | Yes | Cross-referenced with Finance monthly report |

> **Rule:** If you can't fill out a QA log with at least 5 cross-reference checks, you haven't validated the dashboard.

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| "It looks right" | Visual confirmation bias | Always compare to a number from outside the dashboard |
| Testing only the happy path | Time pressure | Test edge cases: empty segments, date boundaries, single-row groups |
| No QA log | "I checked it in my head" | Write it down. Future you (and your manager) will thank you. |
| QA after the stakeholder meeting | "I'll fix it if they notice" | They will notice. QA before every share-out. |

<a id='phase-f-presenting-to-stakeholders'></a>

### F. Presenting to Stakeholders

**Signals you're in this phase:** Dashboard is built and validated. Time to walk the stakeholder through it.

**In plain language:** A dashboard walkthrough is not a tool demo. It's a story. You're not showing them where buttons are — you're showing them what the data says, why it matters, and what they should do next.

---

**Walkthrough Structure (the 4-Part Framework):**

| Part | Duration | What You Say |
|------|----------|-------------|
| **1. Context** | 30 seconds | "This dashboard answers [the question you asked]. It covers [date range] and refreshes [frequency]. Here's how it's organized." |
| **2. Headlines** | 2 minutes | "The three things that jump out: [Metric A] is [up/down] by [X%] vs. [benchmark]. [Metric B] is [trending / holding steady]. [Metric C] is worth watching because [reason]." |
| **3. Drill-down** | 3–5 minutes | Walk through 1–2 interesting breakdowns. Show the filters. Demonstrate how they'd use it independently. |
| **4. Next steps** | 1 minute | "Based on this, I'd recommend [action]. I'll need your input on [decision]. I'll plan to [iterate/hand off/schedule review]." |

---

**Handling "Can you also add X?" in the meeting:**

This will happen. It always happens. Here's your script:

> "That's a great idea. Let me write it down. I want to make sure I deliver what we've scoped here cleanly first, and then I can evaluate adding [request] in the next iteration. Some additions are quick — others might need a different data source or affect performance. I'll follow up with a feasibility assessment."

**Why this works:** It validates the stakeholder, commits to follow-up, but protects the current scope.

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| Showing the tool instead of the story | You're proud of the build | Lead with insight, not navigation |
| No headlines | Expecting the dashboard to speak for itself | Always prepare 2–3 key takeaways before the meeting |
| Saying yes to every add-on request | Wanting to be agreeable | Use the script above. Scope additions for the next iteration. |
| Presenting without QA | Ran out of time | Never present an unvalidated dashboard. Reschedule if needed. |

<a id='phase-g-iteration-and-feedback'></a>

### G. Iteration & Feedback

**Signals you're in this phase:** Stakeholders have seen the dashboard and have feedback. You need to decide what to change, what to defer, and what to push back on.

**In plain language:** Iteration is where dashboards improve — or where scope creep kills them. The skill here is not just executing changes, but *triaging* them.

---

**Feedback Triage Framework:**

| Priority | Criteria | Action | Example |
|----------|----------|--------|---------|
| **P0 — Data is wrong** | A number is incorrect or misleading | Fix immediately before next use | "Churn rate doesn't match our Finance report" |
| **P1 — Missing critical context** | A key metric or comparison is absent | Add in this iteration | "Need to see this by region, not just total" |
| **P2 — Usability improvement** | Layout, labeling, or filter improvements | Add if time allows | "Can the default date range be last 30 days?" |
| **P3 — Nice to have** | New metrics, new views, additional drill-downs | Backlog for future iteration | "Could we also track NPS alongside revenue?" |
| **Out of scope** | Requires new data sources, different tooling, or a fundamentally different dashboard | Decline or scope as a separate project | "Can this also show individual rep performance?" |

---

**Feedback Capture Template:**

> **Dashboard Feedback — [Project Name] — [Date]**
>
> | # | Feedback Item | Source | Priority | Status |
> |---|--------------|--------|----------|--------|
> | 1 | Churn rate off by 0.3% | VP Sales | P0 | Fixed |
> | 2 | Add region breakdown | VP Sales | P1 | In progress |
> | 3 | Change default date range | Product Manager | P2 | Backlogged |
> | 4 | Add NPS data | VP Sales | P3 | Backlogged |
> | 5 | Individual rep view | Sales Director | Out of scope | Separate project |

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| Accepting every request | People-pleasing, wanting to demonstrate value | Triage with the framework above. Share prioritization openly. |
| Changing things without tracking | "Quick fix" that cascades | Log every change. Version your saves. |
| Infinite iteration | No definition of "done" | Agree on iteration count upfront: "Two rounds of feedback, then handoff." |
| Stakeholder adds stakeholders | "Can you also show this to the other team?" | New audience = new scoping conversation |

<a id='phase-h-handoff-and-documentation'></a>

### H. Handoff & Documentation

**Signals you're in this phase:** The dashboard is finalized, validated, and stakeholder-approved. Time to make it self-sustaining.

**In plain language:** A dashboard without documentation is a ticking time bomb. When you leave the team, go on vacation, or just forget the details in 6 months, documentation is the difference between a living tool and a broken artifact.

---

**Documentation Template:**

> **Dashboard Documentation — [Dashboard Name]**
>
> **Purpose:** [One sentence — what decision this supports]
>
> **Owner / Maintainer:** [Name, team, contact]
>
> **Audience:** [Who uses it and how often]
>
> **Data Sources:**
> | Metric | Source Table | Refresh Schedule | Owner |
> |--------|-------------|-----------------|-------|
> | MRR | billing.subscriptions | Daily, 6am UTC | Data Engineering |
> | Active Users | product.user_events | Daily, 7am UTC | Product Analytics |
>
> **Filters & Logic:**
> - Date range default: Last 30 days (rolling)
> - "Active" = logged in within last 30 days, excludes internal accounts
> - Revenue excludes one-time charges and trial accounts
>
> **Known Limitations:**
> - Refund data has 1-day lag
> - Mobile usage not tracked before 2025-06-01
>
> **Change Log:**
> | Date | Change | Reason |
> |------|--------|--------|
> | 03-29-2026 | Initial build | Requested by VP Sales |
>
> **Sunset Criteria:** Review usage quarterly. If < 5 views/month for 2 consecutive quarters, archive.

---

**Access Provisioning Checklist:**

| Item | Done? |
|------|-------|
| Dashboard shared with correct user group |  |
| View-only permissions for consumers, edit for maintainers |  |
| Data source credentials not embedded in dashboard |  |
| Scheduled email/Slack delivery configured (if applicable) |  |
| Bookmark link shared in team channel or wiki |  |

---

**Pitfalls:**

| Pitfall | Why It Happens | How to Avoid It |
|---------|---------------|-----------------|
| No documentation | "I'll remember" | You won't. Write it during build, not after. |
| Documentation as afterthought | "The dashboard is self-explanatory" | It's not. Especially calculated fields and filter logic. |
| No owner assigned | "Everyone owns it" | No one owns it. Assign a specific person. |
| No sunset plan | Dashboards accumulate forever | Set review dates. Archive unused dashboards. |

<a id='4-dashboard-component-pattern-library'></a>

## 4. Dashboard Component Pattern Library

When designing your layout (Phase C), use this reference to pick the right visual component for each metric.

| Component | Best For | Signals to Use It | Common Mistake | Example Context |
|-----------|----------|-------------------|----------------|-----------------|
| **Scorecard / KPI tile** | Headline numbers with comparison | "What is [metric] right now?" | No comparison context (show vs. target or vs. prior period) | Revenue: $2.1M (+8% MoM) |
| **Line chart (time series)** | Trends over time | "How has [metric] changed?" | Too many lines (max 5; use highlighting or small multiples) | Monthly revenue, 12-month trend |
| **Bar chart (vertical)** | Comparing categories | "Which [segment] is highest/lowest?" | Sorting randomly instead of by value | Revenue by product line |
| **Bar chart (horizontal)** | Ranking or long category names | "Top 10 / Bottom 10 of [dimension]" | Showing all 200 categories (limit to top/bottom N) | Top 10 customers by spend |
| **Stacked bar** | Part-to-whole across categories | "What's the composition of [metric] by [segment]?" | Too many segments (max 4–5; group rest as "Other") | Revenue by plan type per quarter |
| **Pie / donut chart** | Part-to-whole (single point in time) | "What percentage is [segment]?" | More than 5 slices, or comparing across time | Current customer mix by plan |
| **Heat map** | Pattern detection across two dimensions | "Where are the hot spots?" | Too granular — hard to read at cell level | Support tickets by hour × day of week |
| **Table with conditional formatting** | Detail-oriented users who want raw numbers | "Show me the actual data" | No formatting — just a wall of numbers | Regional performance summary |
| **Filter bar** | User-driven exploration | "Let me see this for my [region/team/product]" | Too many filters (3–5 max visible; hide the rest) | Region, date range, product line |
| **Sparkline** | Inline trend alongside a KPI | "Is this metric trending up or down?" | Using when absolute values matter more than direction | MoM change indicators in a scorecard |

The tool should match the dashboard type, your organization's stack, and the maintenance model.

|--------|---------|----------------------|----------|---------------|-------|

<a id='6-what-to-avoid'></a>

## 6. What to Avoid — Anti-Patterns

Dashboard projects fail for two reasons: technical mistakes and political mistakes. Both are preventable.

### Technical Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| Unfiltered aggregations | SUM without WHERE includes test data, deleted records, duplicates | Always apply business logic filters before aggregation |
| Hardcoded date ranges | Dashboard breaks on Jan 1 or becomes stale silently | Use relative date functions: LAST 12 MONTHS, YEAR TO DATE |
| Multiple definitions of the same metric | Two charts show different "revenue" numbers | Centralize calculations; one definition, one source |
| No NULL handling | NULLs silently drop from counts and averages | Explicitly handle: COALESCE, IFNULL, or documented exclusion |
| Missing date range in title/header | "Revenue: $2.1M" — for when? | Always show the date range and last-refreshed timestamp |
| Dashboard with 20+ charts | Information overload; nothing stands out | 5–8 charts max. Split into tabs if more views are needed. |
| No loading performance test | Dashboard takes 45 seconds to load | Test with production data volumes. Pre-aggregate heavy queries. |

### Political & Process Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| Building before alignment | You deliver something nobody asked for | Always confirm scope in writing before building |
| Skipping QA | First stakeholder meeting becomes a bug hunt | Validate every number before any share-out |
| Dashboard as email attachment | Screenshot in email = no interactivity, stale immediately | Share a live link. Screenshots are for context in messages, not the deliverable. |
| No documentation or handoff | You become the single point of failure | Write docs during build, not after. Assign a maintainer. |
| Accepting every scope addition | Dashboard becomes a Frankenstein of competing needs | Triage with the feedback framework (Phase G). Say "next iteration." |
| Solving a process problem with a dashboard | "We need visibility" often means "we need a process" | Ask: if this dashboard existed, what would you do differently? If the answer is vague, a dashboard won't help. |
| Presenting without headlines | Stakeholder stares at the screen and says "so... what am I looking at?" | Always prepare 2–3 key takeaways before presenting |
| Building for everyone | "Let's make it for the whole company" | Build for a specific audience. Expand later. |

## 7. Worked Example: Revenue Dashboard for Sales Leadership

This walks through a realistic scenario using the full framework. Each step maps to a phase from Section 3.

---

**Context:** The VP of Sales asks: "I need a dashboard to track our revenue. Can you build something in Tableau?"

### Step-by-Step Walkthrough

| Step | Phase | Action | Output |
|------|-------|--------|--------|
| 1 | **A. Scope** | Schedule 20-min intake meeting. Ask the 7 intake questions. | VP wants to track MRR, churn, and expansion revenue by region. Monthly cadence. Audience: VP + 4 regional directors. Deadline: 2 weeks. |
| 2 | **A. Scope** | Send scope confirmation email using the template. | Written confirmation. VP replies "looks right, but also add Net Revenue Retention." You add it and reconfirm. |
| 3 | **B. Metrics** | Fill out Metric Definition Worksheet for 4 metrics: MRR, Churn Rate, Expansion Revenue, NRR. | Discover that "churn" means different things to Sales (lost logos) vs. Finance (lost dollars). Align on dollar churn with VP. |
| 4 | **B. Data** | Run Data Source Validation Checklist on billing.subscriptions. | Find 47 duplicate rows from a migration bug. Flag to Data Engineering. They fix it. Re-validate. |
| 5 | **2. Classify** | Apply the Decision Tree. Recurring? Yes. Audience? Executives + managers. | Classification: **Strategic dashboard**. Implies: high polish, 4–6 metrics, monthly refresh, 2–3 iteration cycles. |
| 6 | **C. Design** | Sketch wireframe on paper using the Inverted Pyramid layout. | Top row: 4 KPI scorecards (MRR, Churn, Expansion, NRR). Middle: MRR time series (12 months). Bottom left: MRR by region (bar chart). Bottom right: churn by region (bar chart). |
| 7 | **D. Build** | Build in Tableau. Use naming conventions. Create centralized calculated fields. | Working draft in 3 hours. Version saved as `revenue_dashboard_v1_draft`. |
| 8 | **E. QA** | Run the QA Ladder. Fill out the QA Log. | March MRR matches Finance report within $12 (rounding). Churn rate matches after aligning on dollar churn definition. All breakdowns sum to totals. |
| 9 | **F. Present** | 30-min walkthrough with VP using the 4-Part Framework. | VP says: "This is great. Can we also see individual rep performance?" You use the deflection script: "Let me scope that separately." |
| 10 | **G. Iterate** | Triage feedback. VP's "also add renewal rates" is P1. "Rep performance" is out of scope. | Renewal rate added in iteration 2. Rep performance scoped as separate workstream. |
| 11 | **H. Handoff** | Fill out Documentation Template. Share in team wiki. Set up weekly Slack delivery. | Dashboard live. Documentation complete. Quarterly review date set. |

> **Total time:** ~18 hours across 2 weeks. Scoping and metrics (Steps 1–4) took ~5 hours — 28% of total time. This front-loading prevented at least two significant rework cycles.

<a id='9-quick-reference-checklists'></a>

## 9. Quick-Reference Checklists

Use these as-is for your own projects. Copy into your project tracker, Notion page, or personal checklist.

---

### Phase A — Scoping Checklist

- [ ] Problem statement captured in writing
- [ ] Decision the dashboard supports is identified
- [ ] Audience and access list defined
- [ ] Refresh cadence agreed (daily/weekly/monthly/once)
- [ ] Key metrics listed (3–8)
- [ ] Out-of-scope items documented
- [ ] Timeline and deadline confirmed
- [ ] Scope confirmation sent and acknowledged

### Phase B — Metrics & Data Checklist

- [ ] Each metric has a written business definition
- [ ] Source table(s) identified for each metric
- [ ] Metric calculation documented (SQL or pseudocode)
- [ ] Source of truth owner confirmed
- [ ] Row count sanity check passed
- [ ] NULL prevalence checked
- [ ] Data freshness confirmed
- [ ] At least one cross-reference validated against external source

### Phase C — Design Checklist

- [ ] Dashboard type classified (ad-hoc / operational / strategic)
- [ ] Wireframe sketched (paper or digital)
- [ ] Scorecard KPIs selected (3–5)
- [ ] Primary chart identified (the one that answers the main question)
- [ ] Filter set defined with sensible defaults
- [ ] 5-second test: can a glancer tell good vs. bad?
- [ ] Viewing context considered (laptop, projector, mobile)

### Phase D — Build Checklist

- [ ] Naming conventions applied to all calculated fields
- [ ] Date handling consistent (calendar vs. rolling)
- [ ] Calculations centralized (not duplicated per chart)
- [ ] Filters mapped to views (documented which affects which)
- [ ] Performance tested with production data volume
- [ ] Version saved before sharing

### Phase E — QA Checklist

- [ ] Row counts match source
- [ ] Grand totals match SUM() from source query
- [ ] Segment breakdowns sum to grand total
- [ ] Every filter tested
- [ ] Date boundaries verified (first and last dates)
- [ ] NULLs accounted for
- [ ] At least one metric cross-referenced to external source
- [ ] Edge cases tested (zeros, single-row segments)
- [ ] QA log filled out and saved

### Phase F — Presentation Checklist

- [ ] 2–3 key takeaways prepared
- [ ] Walkthrough follows 4-Part Framework (Context → Headlines → Drill-down → Next steps)
- [ ] "Can you also add X?" deflection script ready
- [ ] Dashboard validated before meeting (no last-minute changes)

### Phase G — Iteration Checklist

- [ ] All feedback captured in writing
- [ ] Each item triaged (P0/P1/P2/P3/out of scope)
- [ ] Prioritization shared with stakeholder
- [ ] Iteration count agreed (usually 2 rounds)
- [ ] Changes logged and versioned

### Phase H — Handoff Checklist

- [ ] Documentation template filled out completely
- [ ] Owner/maintainer assigned
- [ ] Access provisioned (correct groups, view-only vs. edit)
- [ ] Scheduled delivery configured (if applicable)
- [ ] Link shared in team channel/wiki
- [ ] Quarterly review date set
- [ ] Sunset criteria documented

<a id='10-final-takeaway'></a>

## 10. Final Takeaway

> **The one sentence to remember:** A great dashboard is not the one with the best charts — it's the one that was scoped correctly, validated rigorously, and documented for the person who comes after you.

---

**The script to internalize:**

*"Before I build: who is this for, what decisions does it support, and what does 'done' look like?*

*During the build: is every number traceable back to a source, and does the layout answer the main question in 5 seconds?*

*After I ship: can someone who wasn't in the room understand how this works, how to maintain it, and when to retire it?"*

---

**What separates good analysts from great ones is not tool mastery — it's the discipline to scope before building, validate before presenting, and document before moving on.**
