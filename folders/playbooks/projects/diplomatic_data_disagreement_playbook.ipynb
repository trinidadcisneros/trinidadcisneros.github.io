# How to Say "The Data Doesn't Support That" Diplomatically

**A professional playbook for navigating disagreements with senior stakeholders when the numbers contradict their intuition or preferred narrative.**

Every analyst will face this moment: a VP makes a confident claim in a meeting, a director builds a strategy around an assumption, or an exec presents a narrative to the board — and your data says they're wrong. Maybe not completely wrong, but wrong enough that it matters.

This is the highest-stakes communication challenge in analytics. Get it wrong and you damage the relationship, get sidelined, or — worst case — stop raising uncomfortable truths altogether. Get it right and you become the person senior leaders trust precisely *because* you're willing to push back with evidence and tact.

This guide covers how to read the room before you speak, how to frame data-driven disagreements so they land as insights rather than attacks, how to handle the emotional dynamics, and how to recover when it doesn't go well.

**How to use this guide:**
- **In the moment:** Jump to the Response Framework (Section 3) for ready-to-use language
- **Before a tough meeting:** Read Sections 2 and 4 for preparation tactics
- **Building the habit:** Review the Anti-Patterns (Section 7) and Worked Examples (Section 8)
- **After a rough conversation:** Section 6 covers recovery and repair

## 1. Why This Is Hard — The Dynamics at Play

Data disagreements with senior stakeholders aren't just about numbers. They're about power, identity, and trust — and if you don't understand those dynamics, even perfect data won't save you.

| Dynamic | What's Really Happening | Why It Matters for You |
|---------|------------------------|------------------------|
| **Seniority gap** | They have more organizational power than you | Being "right" doesn't matter if you make them feel undermined in front of their peers |
| **Narrative investment** | They've already told a story — to their boss, their team, maybe the board | Contradicting the data contradicts *them*, and they may have already staked credibility on it |
| **Intuition vs. evidence** | Their gut feeling comes from years of experience you don't have | Dismissing intuition as "wrong" dismisses their expertise. Sometimes intuition is picking up something the data hasn't captured yet |
| **Public vs. private** | Being corrected in a meeting feels different than in a 1:1 | A stakeholder who would thank you in private might fight you in public |
| **Messenger risk** | Analysts who only bring bad news get labeled as blockers | You need to be seen as someone who helps, not someone who pokes holes |

---

### The Core Tension

Your job is to tell the truth. Your career depends on maintaining relationships. These two things can feel contradictory, but they're not — if you have the right approach.

The goal is never to "win" the argument. The goal is to **make sure the best available information reaches the decision**. Sometimes that means changing someone's mind. Sometimes it means planting a seed. Sometimes it means documenting your concern and letting the decision-maker decide.

> **The mindset to adopt:** "I'm not here to prove you wrong. I'm here to make sure we're all working from the same facts so we make the best decision."

## 2. Before You Speak — The Readiness Checklist

Before you challenge a stakeholder's claim, make sure you've earned the right to do so. Rushing to correct someone with half-checked data is worse than saying nothing.

---

### The 5-Point Readiness Check

| # | Question | If No, Then... |
|---|---------|----------------|
| 1 | **Is your data solid?** Have you QA'd the numbers, cross-referenced at least one, and confirmed your methodology? | Stop. QA first. Being wrong while correcting someone is career-damaging. |
| 2 | **Do you understand their claim?** Can you state their position charitably — the strongest version of what they believe? | Clarify first. You might be disagreeing with something they didn't actually say. |
| 3 | **Is the gap material?** Does the difference between their claim and the data actually change the decision? | Let it go. Not every inaccuracy needs correcting. Pick your battles. |
| 4 | **Have you considered what the data might be missing?** Could their intuition be capturing something your analysis doesn't? | Acknowledge the gap. "The data shows X, but it doesn't capture Y, which might explain your read." |
| 5 | **Do you have a constructive frame?** Can you present this as adding information, not as tearing down their position? | Reframe before speaking. "Here's what I found" beats "That's not what the data says." |

> **The rule:** If you can't pass all 5 checks, you're not ready to push back yet. Do the prep work first.

---

### The Charitable Interpretation Exercise

Before you disagree, write down (even mentally) the strongest version of their argument. This does two things: it prevents you from arguing with a straw man, and it shows respect when you reference it out loud.

| What They Said | Uncharitable Interpretation | Charitable Interpretation |
|---------------|---------------------------|-------------------------|
| "Our customers love the new feature" | They're cherry-picking feedback | They've been hearing positive signals from key accounts that may not show up in aggregate metrics yet |
| "Churn isn't a problem" | They're ignoring the data | They may be looking at a different cohort, time window, or definition of churn than you are |
| "We should double down on enterprise" | They're biased toward big deals | They may have context about pipeline or strategic conversations you're not privy to |
| "The campaign was a huge success" | They're inflating results | They may be measuring a different KPI or comparing to a different baseline than your analysis used |

**Start with the charitable interpretation.** If it turns out they really are just wrong, you'll still get to that conclusion — but you'll get there in a way that doesn't burn the relationship.

## 3. The Response Framework — How to Frame It

The difference between "the data says you're wrong" and "I found something that might change the picture" is entirely in the framing. Same data, different outcome.

---

### The A.D.D. Framework (Acknowledge → Data → Door)

This is your go-to structure for any data-driven disagreement:

| Step | What You Do | Why It Works |
|------|------------|-------------|
| **Acknowledge** | Validate their perspective or the logic behind their position | Shows you're not dismissing them. Lowers defenses. |
| **Data** | Share what the data shows — as a discovery, not a correction | Positions you as bringing new information, not contradicting them |
| **Door** | Open a path forward — invite collaboration on interpreting the data together | Keeps it a conversation, not a verdict. Gives them room to adjust without losing face |

---

### A.D.D. in Practice — The Phrase Library

#### Acknowledge Phrases

| Situation | What to Say |
|-----------|------------|
| They made a claim you disagree with | "That's an interesting read — and I can see why that'd be the takeaway from [what they saw]." |
| They stated something as fact | "That lines up with what we were seeing earlier in the year..." |
| They're passionate about a direction | "I think the instinct behind that is right — [restate the underlying goal they're trying to achieve]." |
| They're in front of their team/leadership | "You've been closer to this than anyone, so I want to share something I noticed to make sure we have the full picture." |

#### Data Phrases

| Situation | What to Say |
|-----------|------------|
| Direct contradiction | "When I pulled the numbers for [time period], I'm seeing a different pattern..." |
| Partial contradiction | "The overall trend supports that — but when I break it down by [segment], there's a wrinkle worth flagging." |
| Their number is wrong | "I pulled from [source] and got [number]. It might be a timing or filter difference — worth reconciling so we're all working from the same baseline." |
| The conclusion doesn't follow | "The metric did go up, but when I controlled for [factor], it looks like most of that lift came from [alternative explanation]." |

#### Door Phrases

| Situation | What to Say |
|-----------|------------|
| You want to continue the conversation | "I'd love to dig into this more — can we look at it together and see if there's something the data might be missing?" |
| You want them to reconsider | "Would it be worth running a quick analysis on [their assumption] to pressure-test it before the decision?" |
| You want to move on but document the gap | "I'll send over what I found so you have it for reference. If the picture changes, happy to revisit." |
| You want to propose an alternative | "What if we looked at it from [different angle]? That might reconcile what you're seeing with what the data shows." |

---

### Full A.D.D. Examples

**Scenario:** VP says "The new onboarding flow is working great — activation is up 15%." Your data shows activation is up 15% overall but the new flow cohort is actually *lower* than the control group; the lift is from a seasonal effect.

> **Acknowledge:** "I saw that 15% lift too — that's a great top-line result and the team should feel good about shipping it."
>
> **Data:** "When I split the cohorts, though, I noticed something interesting. Accounts that went through the new flow activated at 38%, while the legacy flow cohort activated at 41%. The overall 15% lift looks like it's mostly seasonal — we see a similar bump this time every year."
>
> **Door:** "It might be worth running a controlled comparison for another 2-3 weeks to isolate the onboarding effect from the seasonal one. Want me to set that up?"

**Scenario:** Director says "We need to hire 3 more SDRs — our pipeline coverage is too thin." Your data shows pipeline coverage is actually at target, but conversion rate dropped, which is a different problem.

> **Acknowledge:** "I totally get the concern — when deals aren't closing at the rate we need, it's natural to look at the top of the funnel."
>
> **Data:** "The pipeline coverage ratio is actually at 3.2x, which is right at our target. Where I see the gap is conversion: stage 2 to stage 3 conversion dropped from 35% to 22% last quarter. So we have enough pipeline, but we're losing more of it mid-funnel."
>
> **Door:** "Would it make sense to look at what's happening at that stage before deciding on the hiring plan? If the conversion drop is fixable, we might not need the headcount."

## 4. Reading the Room — When to Push and When to Pause

Not every situation calls for the same level of pushback. The art is knowing when to speak up immediately, when to take it offline, and when to let it go.

---

### The Decision Matrix

| | **Low stakes** (the error won't change the decision) | **High stakes** (the error could lead to a bad decision) |
|---|---|---|
| **Private setting** (1:1 or small group) | Mention it casually. Low risk, low reward. | Raise it clearly using A.D.D. This is the ideal scenario. |
| **Public setting** (large meeting, exec review) | Let it go. Not worth the risk. Note it for a follow-up. | Flag it gently now, then follow up in detail afterward. |

### The "Public Meeting" Playbook

When a senior leader makes a data claim in a large meeting that you know is wrong, here are your options ranked from safest to most assertive:

| Level | What You Do | Example Language |
|-------|------------|------------------|
| **1. Seed a question** | Ask a question that surfaces the discrepancy without stating it | "I'd love to understand that better — are we looking at the same cohort? I want to make sure I'm pulling from the right baseline." |
| **2. Offer to validate** | Volunteer to check, implying there might be a gap | "That's a great point. I'd like to cross-reference that with what I'm seeing in [source] — mind if I follow up with an update after the meeting?" |
| **3. Gentle flag** | State the discrepancy softly, leave room for you to be wrong | "I was looking at this recently and got a slightly different number — could be a timing or filter thing. Let me reconcile and share what I find." |
| **4. Direct but diplomatic** | State the data clearly using A.D.D., but only if the decision is about to be made *now* | "I want to flag something before we commit — when I ran the numbers this morning, I saw [different finding]. It might change the picture on [decision]." |

> **The principle:** In public settings, prioritize the *relationship* in the room and the *data* in the follow-up. You can always share findings after the meeting. You can't un-embarrass someone.

## 5. The Escalation Ladder — From Gentle Nudge to Formal Escalation

Not every disagreement gets resolved in one conversation. Sometimes you'll need to escalate — but escalation has a specific order, and skipping rungs damages trust.

---

### The 5 Rungs

| Rung | Action | When to Use | Example |
|------|--------|------------|--------|
| **1. Informal mention** | Raise it casually in a 1:1 | First time you notice the discrepancy; low urgency | "Hey, I noticed something in the data that might nuance the churn story — want me to pull it together?" |
| **2. Structured share** | Send a written analysis using A.D.D. (email or doc) | You've done the homework and have clear findings | "I put together a quick analysis — here's what I'm seeing and why I think it matters for the decision." |
| **3. Explicit disagreement** | State clearly, in private, that you believe the current direction is based on inaccurate data | The stakeholder saw your analysis and dismissed it without engaging | "I want to be transparent — I'm concerned the data we're using to make this call isn't painting the full picture." |
| **4. Bring in a third party** | Loop in your manager, their peer, or a shared stakeholder to weigh in | The disagreement is stuck and the stakes are high | "I've shared my concerns with [stakeholder] and we see it differently. I'd value your perspective before the team commits." |
| **5. Formal documentation** | Put your analysis and concerns in writing, share with decision-makers, and let them decide | You've exhausted dialogue; the decision is imminent and the risk is significant | "I want to make sure my findings are on record. Here's the analysis. I understand the final call is yours, and I'll support whatever direction we go." |

---

### Rules of Escalation

| Rule | Why |
|------|-----|
| Never skip rungs | Going from rung 1 to rung 4 looks like you're going over someone's head. Always give the person a chance to engage first. |
| Always escalate privately first | Even rung 5 should be preceded by a private heads-up: "I'm going to share my analysis with the group — wanted you to see it first." |
| Don't escalate to "win" | You escalate because the *decision quality* is at risk, not because your ego needs validation. If you're motivated by being right, pause. |
| Know when to stop | If you've reached rung 5 and the decision-maker still disagrees, document your position and move on. You've done your job. The decision is theirs. |
| Separate the analysis from the relationship | After the decision is made, regardless of outcome, re-engage the stakeholder professionally. Don't hold a grudge. |

## 6. When It Goes Sideways — Recovery Tactics

Sometimes, despite your best efforts, the conversation goes badly. The stakeholder gets defensive, dismisses the data, or the room gets uncomfortable. Here's how to recover.

---

### Scenario-Based Recovery

| What Happened | What to Do Next | Example Language |
|---------------|----------------|------------------|
| **They got defensive in the meeting** | Don't push further in the moment. Let it cool. Follow up in private within 24 hours. | "Hey, I didn't mean to put you on the spot earlier. I've got some data I think is worth looking at — can I walk you through it 1:1?" |
| **They dismissed the data outright** | Send the analysis in writing so it's on record. Don't force a response. | "I put together a short write-up of what I found. No pressure to respond — just wanted you to have it in case it's useful." |
| **They questioned your competence** | Stay calm. Don't defend yourself in the room. Let the data speak for itself in the follow-up. | "That's fair — let me double-check my methodology and come back to you with a tighter version. I want to make sure I'm giving you the best information." |
| **The room sided with them despite the data** | Accept it gracefully. Document your findings. Revisit only if outcomes prove you right. | "Understood. I'll note my findings and we can revisit if the results come in differently than expected." |
| **You realized mid-conversation that your data was wrong** | Own it immediately. Don't try to save face. | "Wait — I think I might have an error in my filter. Let me re-pull this before we go any further. I don't want to waste your time with bad numbers." |

---

### The 48-Hour Rule

After a difficult data disagreement, follow up within 48 hours with one of these messages:

**If you were right (and they know it):**
> "I know that was an uncomfortable conversation, and I appreciate you hearing me out. I want you to know I'm not trying to be a contrarian — I just want to make sure we have the full picture. Let me know if there's anything I can do to support the next step."

**If you were wrong:**
> "I went back and rechecked — you were right about [specific thing]. I had [explain the error]. I appreciate your patience, and I'll make sure to double-check that going forward."

**If it's still unresolved:**
> "I've been thinking about our conversation and I still see the data differently — but I also recognize you have context I might be missing. Would you be open to sitting down for 20 minutes to walk through it together? I think we can get aligned."

---

### Protecting Your Reputation Long-Term

One bad interaction won't define you. What defines you is the pattern. Here's how to build a reputation as someone who pushes back *well*:

| Do This Consistently | Why It Matters |
|---------------------|----------------|
| Bring good news too, not just corrections | If you only show up when someone is wrong, you'll be seen as a critic. Share wins and positive findings just as often. |
| Be right more often than you're wrong | Every time you push back and the data is solid, you build credibility. Every time you push back and you're wrong, you spend it. |
| Show your work | When people can see your methodology, they trust your conclusions — even uncomfortable ones. |
| Give credit when their intuition was right | "Your instinct was exactly right — the data confirms it." This builds the account you'll withdraw from when you disagree. |
| Never say "I told you so" | Even when results prove you were right. Especially then. |

## 7. What to Avoid — Anti-Patterns

An **anti-pattern** is a common response that feels natural or intuitive in the moment but consistently leads to bad outcomes. These are habits worth naming so you can catch yourself before they do damage.

### Communication Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| "Actually, the data says..." | Opens with a correction. Puts them on the defensive immediately. | "I found something interesting when I looked into this..." |
| "You're wrong about..." | Personal. Confrontational. Career-limiting. | "The data tells a slightly different story..." |
| "The data clearly shows..." | Implies their position is irrational. Nobody likes being told the answer is "clear" when they disagree. | "What I'm seeing in the data is..." |
| Correcting them in front of their boss | Maximum embarrassment, maximum retaliation. | Take it offline. Always. |
| CC'ing leadership on the correction email | Goes over their head. Destroys trust permanently. | Send to them first. Let them loop in others if needed. |
| Repeating your point louder when they push back | Escalating volume or frequency doesn't make data more persuasive. It makes you annoying. | State your case once, clearly. Then listen. |
| Using jargon to establish authority | Saying "the p-value is 0.03" to a non-technical VP doesn't persuade — it intimidates or alienates. | Translate: "There's strong statistical evidence that this difference is real, not random noise." |
| Flooding them with charts and data | Overwhelming someone with evidence makes them feel ambushed, not informed. | Lead with one clear finding. Offer the supporting data if they want it. |

### Strategic Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| Picking every battle | If you correct every minor claim, people stop listening. You become noise. | Save your credibility for the disagreements that actually affect decisions. |
| Never picking any battle | If you never push back, you're not adding analytical value. You're a report generator. | The goal is to push back when it matters, with evidence and empathy. |
| Making it about being right | The moment it becomes about your ego, you've lost — even if the data is on your side. | Make it about the decision quality, not about who's correct. |
| Assuming malice | Most stakeholders aren't lying or manipulating data. They're working from different information or a different lens. | Start with curiosity: "Help me understand how you got to that conclusion." |
| Going silent when overruled | Sulking or withdrawing after being overruled looks petty and unprofessional. | Document your position, support the decision, and revisit if outcomes warrant it. |

## 8. Worked Examples

### Example 1: The Exec Who Loves a Narrative

**The situation:** The CMO is presenting to the board next week. Her narrative: "Our brand campaign drove a 25% increase in inbound leads." You've looked at the data and the lead increase started *before* the campaign launched and correlates more closely with a seasonal trend and a pricing change.

**Why this is hard:** She's already built her board deck around this narrative. She's excited. Her team worked hard on the campaign.

**What NOT to do:**
- Send an email saying "The campaign didn't actually drive leads."
- Present your analysis in the next team meeting.
- CC her boss on your findings.

**What to do:**

**Step 1 — Private message (Slack or in-person):**
> "Hey, I've been digging into the lead data to support your board prep, and I noticed something that's worth discussing before the presentation. Can I grab 15 minutes with you?"

**Step 2 — In the 1:1, use A.D.D.:**
> **Acknowledge:** "The 25% lift is real — that's a strong result and the board will react well to it."
>
> **Data:** "When I looked at the timing, though, the increase started about two weeks before the campaign launched. There's a seasonal pattern from last year that matches closely, and the pricing change we made in the same window also correlates. I'm not saying the campaign didn't contribute — it might have amplified the trend — but attributing the full 25% to it could be risky if someone on the board asks about the timeline."
>
> **Door:** "What if we frame it as 'the campaign, combined with favorable market timing and our pricing adjustment, drove a 25% increase in leads'? That's still a strong story, and it's defensible if someone digs into the data. I can also help you build a slide that shows the contribution breakdown, if that's useful."

**Why this works:** You protected her from a board-level embarrassment. You gave her a way to keep the win. You offered to help rather than just critique. She'll remember this the next time she's building a narrative — and she'll come to you *before* the deck is done.

---

### Example 2: The VP Who Doesn't Believe the Numbers

**The situation:** The VP of Sales insists that enterprise win rates are "at least 40%." Your CRM data shows 28%. He's basing his feeling on the deals he's personally involved in (which are, in fact, around 40% — because they're the biggest, most-resourced deals).

**Why this is hard:** He's not wrong about *his* experience. He's wrong about the aggregate. And he's making hiring and quota decisions based on 40%.

**What to do:**

**Step 1 — Lead with his truth:**
> "You're absolutely right that the deals you're closest to are closing at around 40% — I confirmed that. Those tend to be our largest, most strategic deals with full executive sponsorship."

**Step 2 — Introduce the bigger picture:**
> "When I look at all enterprise deals in the pipeline — including the ones managed by individual reps without exec sponsorship — the blended rate comes in at 28%. The gap is mostly in the mid-market enterprise segment where deals are smaller and reps are running them more independently."

**Step 3 — Make it useful, not threatening:**
> "This actually tells a really positive story about executive involvement. When you or your leaders are engaged, win rates are 12 points higher. If we could replicate even part of that engagement model across the broader pipeline — maybe through a structured deal review cadence — we might be able to close the gap. Want me to model what that would look like for quota planning?"

**Why this works:** You validated his experience, explained the discrepancy without making him wrong, and turned the finding into an opportunity that benefits his team. He's now invested in the correct number because it supports a strategy he can own.

---

### Example 3: The Decision That's Already Been Made

**The situation:** Leadership has decided to sunset a product feature. The rationale presented: "Usage is low — only 8% of accounts use it." You've looked at the data and found that the 8% represents the company's highest-value enterprise accounts, and the feature is cited as a top-3 reason for choosing the product in win/loss interviews.

**Why this is hard:** The decision is made. Leadership has moved on. Reopening it makes you look like you can't accept a decision.

**What to do:**

**Step 1 — Acknowledge the decision:**
> "I understand the reasoning behind sunsetting [feature] — 8% adoption does look low, and the team needs to focus resources."

**Step 2 — Share the risk, not the disagreement:**
> "I wanted to flag a risk I found in the data: the 8% of accounts using this feature represent 34% of our ARR. They're disproportionately our largest enterprise customers. I also checked the win/loss data, and this feature is mentioned in 3 of the top 5 reasons enterprise accounts chose us over competitors."

**Step 3 — Offer a middle path:**
> "I'm not suggesting we reverse the decision — just that we might want to mitigate the risk. Would it make sense to do targeted outreach to the affected accounts before the sunset? Or phase the deprecation to give them a migration path? I can model the revenue at risk if that's helpful."

**Why this works:** You respected the decision. You didn't say "you're wrong" — you said "here's a risk." You offered solutions, not just problems. And you gave them a way to adjust without admitting they missed something.

## 9. Quick-Reference Checklists

### Before Raising a Data Disagreement

- [ ] Data is QA'd — you'd bet real money on it
- [ ] You can state their position charitably (strongest version)
- [ ] The gap is material — it changes the decision
- [ ] You've considered what the data might be missing
- [ ] You have a constructive frame (adding information, not tearing down)
- [ ] You've chosen the right setting (private > public)
- [ ] You've assessed timing (is the decision imminent or is there runway?)

### The A.D.D. Quick Check

- [ ] **Acknowledge:** Did you validate their perspective or logic before presenting data?
- [ ] **Data:** Are you presenting findings as a discovery, not a correction?
- [ ] **Door:** Did you open a collaborative path forward?

### After a Difficult Conversation

- [ ] Follow up within 48 hours (regardless of outcome)
- [ ] If you were right: be gracious, not gloating
- [ ] If you were wrong: own it fast and publicly
- [ ] If unresolved: offer to meet and look at it together
- [ ] Document your findings in writing (email or shared doc)
- [ ] Re-engage professionally — don't let one disagreement define the relationship

### Reputation Maintenance

- [ ] Bringing good news and confirmations, not just corrections
- [ ] Giving credit when their intuition is right
- [ ] Showing your work so people trust the methodology
- [ ] Never saying "I told you so" — even when results prove you right
- [ ] Supporting decisions even when you disagree, after stating your case

## 10. Final Takeaway

> **The one sentence to remember:** Your job isn't to prove anyone wrong — it's to make sure the best available information reaches the decision.

---

**The script to internalize:**

*"Before I speak: is the data solid, is the gap material, and do I have a constructive frame?*

*When I speak: acknowledge their view, share what the data shows, and open a door for collaboration.*

*After I speak: follow up, stay professional, and let the decision-maker decide."*

---

**Why this matters:** The analyst who never pushes back is a report generator. The analyst who pushes back poorly is a liability. The analyst who can diplomatically surface uncomfortable truths — with empathy, evidence, and a constructive frame — becomes the person senior leaders trust with the hardest problems. That's the analyst who gets the seat at the table, not because they were the loudest, but because they made every decision better.
