# How to Scope and Deliver an Ad-Hoc Analysis

**A professional playbook for turning vague requests into polished, trusted deliverables.**

Ad-hoc analysis is the most common — and most underestimated — thing analysts do. It's the "can you pull this," the "what's going on with churn," the "I need some numbers for a meeting." Most of these requests arrive incomplete, ambiguous, or solving the wrong problem entirely. The analyst who can consistently translate a vague Slack message into a clear, trustworthy deliverable is the one stakeholders rely on, promote, and fight to keep on their team.

This guide covers the full arc: how to decode what someone is really asking, how to scope the work before you start querying, how to structure your analysis for clarity, how to present findings so they land, and how to avoid the traps that make ad-hoc work feel endless and thankless.

**How to use this guide:**
- **Quick reference:** Jump to the Request Decoder (Section 2) or Deliverable Format Picker (Section 5)
- **Scoping help:** Read Section 3 for the 5-minute scoping framework
- **Structure:** Section 4 gives you the analysis structure template
- **Communication:** Section 7 has response templates for every stage
- **Mistake prevention:** Review Anti-Patterns (Section 8) before delivering

## 1. The Ad-Hoc Lifecycle at a Glance

Every ad-hoc analysis — from a 30-minute data pull to a week-long deep dive — follows the same lifecycle. The difference between good and great analysts is how much discipline they apply to each phase.

| Phase | Key Action | Time Budget | Mistake If Skipped |
|-------|-----------|-------------|-------------------|
| **1. Decode** | Understand what they're really asking | 10% | You answer the literal question, not the real one |
| **2. Scope** | Confirm the question, audience, format, and timeline | 10% | You build the wrong thing or over-build for a quick question |
| **3. Explore** | Pull initial data, check feasibility, spot surprises | 20% | You commit to a deliverable before knowing if the data supports it |
| **4. Analyze** | Build the analysis, find the answer | 30% | — |
| **5. QA** | Validate every number before sharing | 15% | First thing the stakeholder finds is wrong. Trust gone. |
| **6. Deliver** | Send the deliverable with context and a recommendation | 10% | You send a spreadsheet with no narrative. Stakeholder says "so what?" |
| **7. Follow up** | Check if it answered the question, capture learnings | 5% | Missed opportunity to build the relationship and improve your process |

> **The 20% rule:** If you spend less than 20% of your time on Decode + Scope (phases 1–2), you're almost certainly answering the wrong question.

## 2. The Request Decoder — "What Are You Really Asking?"

Most ad-hoc requests have three layers:

| Layer | What It Is | Example |
|-------|-----------|---------|
| **Surface request** | The literal words they used | "Can you pull churn numbers?" |
| **Real question** | What they actually need to know | "Is churn getting worse, and if so, where?" |
| **Underlying decision** | What they'll do with the answer | "Should we invest in retention for enterprise accounts?" |

Your job is to get to the third layer before you start any work. The surface request is almost never the full picture.

---

### The Translation Table

| They Say | They Probably Mean | What to Ask |
|----------|-------------------|-------------|
| "Can you pull [metric]?" | "I need to know if [metric] is good or bad" | "What are you comparing it to? Last month? A target? A competitor?" |
| "What's going on with [thing]?" | "Something seems off and I need to understand why" | "What triggered this? Did you see something specific that concerned you?" |
| "I need some numbers for a meeting" | "I need to look credible in front of [audience]" | "Who's the audience, what's the meeting about, and what point are you trying to make?" |
| "Can you look into [topic]?" | "I have a hypothesis and want data to test it" | "What's your hypothesis? What would change your mind?" |
| "How are we doing on [thing]?" | "I need a status update I can share with my boss" | "What does 'good' look like? Do you have a target or benchmark?" |
| "This doesn't look right" | "The number I see doesn't match my expectation" | "What number did you expect, and where did you see the one that looks off?" |
| "Can you build a model for [thing]?" | Could mean anything from a spreadsheet to a machine learning pipeline | "When you say 'model,' what output do you imagine? A forecast? A score? A recommendation?" |

## 3. The 5-Minute Scoping Framework

Once you've decoded the real question, confirm scope before starting. This takes 5 minutes and prevents hours of rework.

---

### The 6 Questions to Nail Down

| # | Question | Why | Example Answer |
|---|----------|-----|---------------|
| 1 | **What's the question in one sentence?** | Forces clarity for both of you | "Is enterprise churn trending up, and if so, which segments?" |
| 2 | **What decision does this inform?** | Prevents analysis that goes nowhere | "Whether to invest in enterprise retention vs. growth" |
| 3 | **Who's the audience?** | Determines depth, format, and language | "VP of CS, who will share it with her team" |
| 4 | **What time period?** | Prevents "can you go back further?" after delivery | "Last 12 months, compared to same period prior year" |
| 5 | **What format do you need?** | Prevents over- or under-building | "A few charts and a summary I can paste into my deck" |
| 6 | **When do you need it?** | Determines depth vs. speed trade-off | "Thursday EOD for Friday's leadership meeting" |

---

### The Scope Confirmation (30-second Slack message)

After asking these questions, send a quick confirmation:

> "Just to make sure I'm building the right thing:
>
> **Question:** Is enterprise churn trending up, and which segments are driving it?
> **Time period:** Last 12 months vs. prior year
> **Deliverable:** Charts + summary bullets (deck-ready)
> **ETA:** Thursday EOD
>
> Anything I'm missing?"

This is fast, lightweight, and saves you from delivering the wrong analysis. Even if the stakeholder's answer is just "yep looks good," you now have written confirmation of scope.

---

### The Depth-vs-Speed Decision

Not every request deserves a deep dive. Match your effort to the stakes:

| Signal | Depth Level | Effort | Output |
|--------|------------|--------|--------|
| "Quick question" / "just curious" / no deadline | **Quick pull** | 30 min – 2 hours | Number + one line of context in Slack |
| "For a meeting" / "need to present this" | **Structured summary** | 2–8 hours | Charts + narrative + key takeaways |
| "We need to make a decision on..." / exec-facing | **Full analysis** | 1–3 days | Complete write-up: context, methodology, findings, recommendation |
| "This could change our strategy" / board-level | **Deep dive** | 3–10 days | Comprehensive analysis with scenarios, sensitivity, and appendix |

> **The trap to avoid:** Doing deep-dive-level work for a quick-pull-level question. If someone asks "what was last month's revenue?" they don't need a 10-page analysis. Give them the number.

## 4. Structuring the Analysis

Once you're past scoping, you need a structure for the work itself. Without one, ad-hoc analysis becomes a wandering exploration that produces interesting charts but no clear answer.

---

### The SCR Framework (Situation → Complication → Resolution)

This is the backbone of every well-structured analysis. Stolen from consulting, adapted for analytics:

| Element | What It Is | Example |
|---------|-----------|---------|
| **Situation** | The context everyone agrees on | "We've been tracking enterprise churn monthly. The target is < 3% annualized." |
| **Complication** | What changed or what's surprising | "Enterprise churn has risen to 4.8% annualized over the last two quarters, driven primarily by mid-market accounts in the 50–200 employee range." |
| **Resolution** | Your finding and recommendation | "The driver is a pricing mismatch: mid-market accounts on the legacy plan churn at 2x the rate of those on the new plan. Migrating legacy accounts could reduce enterprise churn by ~1.2 percentage points. Recommend piloting migration with the top 20 at-risk accounts." |

---

### The Analysis Work Plan

Before you start querying, write yourself a 5-line work plan. This keeps you from going down rabbit holes:

| Step | What | Time Box |
|------|------|----------|
| 1 | Pull the core metric (churn by segment, 12 months) | 1 hour |
| 2 | Break down by likely drivers (plan type, company size, tenure) | 1.5 hours |
| 3 | Test the leading hypothesis (is one segment driving the trend?) | 1 hour |
| 4 | QA the numbers against a known source | 30 min |
| 5 | Write up findings in SCR format | 1 hour |

**Total time-boxed:** 5 hours. If you hit hour 5 and don't have a clear answer, stop and assess — don't keep digging.

---

## 5. Choosing the Right Deliverable Format

The format should match the audience and the stakes. Over-formatting a quick pull wastes your time. Under-formatting an exec deliverable loses credibility.

| Format | When to Use | Includes | Time to Produce |
|--------|------------|----------|-----------------|
| **Slack message** | Quick pull, single number, low stakes | Number + 1-2 lines of context | 5–15 min (after analysis) |
| **Email with inline findings** | Structured summary, 2–5 key points | Situation, key findings, recommendation, next steps | 30–60 min |
| **Slide-ready charts** | Stakeholder will put it in a presentation | 2–4 charts with titles and takeaway annotations | 1–2 hours |
| **Written memo / doc** | Full analysis, exec-facing, decision-driving | SCR structure, methodology, charts, appendix | 2–4 hours |
| **Interactive dashboard** | Request will recur or audience wants self-service | Live filters, multiple views | Half day+ (see Dashboard Playbook) |

### Format-Specific Tips

**Slack message — the anatomy of a good quick response:**

> [Stakeholder name] — here's what I found:
>
> **Enterprise churn (Mar 2026):** 4.8% annualized, up from 3.1% in Mar 2025
> **Key driver:** Mid-market accounts (50–200 employees) on the legacy plan churn at 2x the rate of new-plan accounts
> **Caveat:** This excludes 12 accounts that are in active renewal negotiation
>
> Want me to dig deeper into the legacy plan angle, or is this enough for your purposes?

This takes 5 minutes to write and gives the stakeholder everything they need to act. Notice: number, comparison, context, caveat, and a clear next step.

**Email with findings — always include these 4 sections:**
1. **Bottom line** (one sentence answer)
2. **Key findings** (3–5 bullets, most important first)
3. **Methodology note** (one line: what data, what time period, what exclusions)
4. **Recommended next step** (what should the reader do with this information?)

## 6. QA Before You Send — The Trust Checklist

Every deliverable, no matter how small, should pass these checks before you hit send. This takes 5–10 minutes and prevents the most common credibility-destroying mistakes.

---

### The Ad-Hoc QA Ladder

| Step | Check | How |
|------|-------|-----|
| 1 | **Does the number pass the smell test?** | Is it in the right order of magnitude? If revenue was 2M last month, is this month's number plausible? |
| 2 | **Are the date boundaries correct?** | Does "last 12 months" actually start 12 months ago? Is the end date inclusive? |
| 3 | **Are NULLs accounted for?** | Did you COUNT(*) vs COUNT(field)? Are NULLs excluded intentionally? |
| 4 | **Can you cross-reference one number?** | Check at least ONE number against a known source (a finance report, a dashboard, a prior deliverable) |
| 5 | **Are filters applied correctly?** | If you're looking at "enterprise only," did you verify the filter worked? |
| 6 | **Is there comparison context?** | Every number needs a "vs. what?" — prior period, target, benchmark |
| 7 | **Did you state your assumptions?** | What's included, what's excluded, what definitions did you use? |
| 8 | **Would you bet 100 dollars this is right?** | If your gut says "probably" — check again. If "yes" — send it. |

> **The rule:** The five minutes you spend on QA are worth more than the five hours you spent on the analysis. One wrong number erases all the right ones.

---

### The "Known Unknowns" Section

Every deliverable should include a brief note on what you know you don't know:

> **Caveats / known limitations:**
> - Excludes 12 accounts currently in renewal negotiation (status = 'pending')
> - Mobile usage data not available before June 2025
> - "Active" defined as logged in within last 30 days — some teams use 7 days

This isn't a weakness — it's a strength. It shows you thought carefully about the data's limitations and saves the stakeholder from misinterpreting results.

## 7. Communication Templates

---

### Template 1: Clarifying a Vague Request

> "Thanks for sending this over! I want to make sure I answer the right question. A few quick clarifications:
>
> - When you say [vague term], do you mean [interpretation A] or [interpretation B]?
> - What decision will this inform? That'll help me figure out what level of detail you need.
> - Any specific time period or segments you care about most?
>
> Happy to jump on a 5-minute call if it's easier to talk through."

---

### Template 2: Scope Confirmation

> "Here's what I'm planning to deliver — want to confirm before I start:
>
> **Question:** [One sentence]
> **Time period:** [Range]
> **Segments/filters:** [What's included and excluded]
> **Format:** [Slack message / email / charts / memo]
> **ETA:** [Date]
>
> Anything off?"

---

### Template 3: Delivering Results (Slack)

> "[Name] — here's what I found:
>
> **Bottom line:** [One sentence answer to the original question]
>
> **Key findings:**
> - [Finding 1 — most important]
> - [Finding 2]
> - [Finding 3]
>
> **Caveat:** [One line on what's excluded or uncertain]
>
> **Recommendation:** [What I suggest based on the data]
>
> Let me know if you want me to dig deeper on any of this."

---

### Template 4: Delivering Results (Email)

> **Subject: Analysis — [Topic] — [One-line finding]**
>
> Hi [Name],
>
> **Bottom line:** [One sentence — the answer to their question]
>
> **Key findings:**
> 1. [Most important finding with the number]
> 2. [Second finding]
> 3. [Third finding]
>
> **Methodology:** This analysis covers [time period], using [data source], filtered to [criteria]. [One-line note on what's excluded.]
>
> **Recommendation:** Based on these findings, I'd suggest [action]. [One sentence on why.]
>
> **Further investigation (if warranted):** [One-liner on what you could dig into next, if the stakeholder wants more]
>
> Happy to walk through this live if helpful. Let me know."

---

### Template 5: Setting Expectations on Timeline

> "Got it — this is a good question. Based on the scope, here's what I'm thinking:
>
> - **Quick version** (top-line numbers, minimal breakdown): I can have this by [date]
> - **Full version** (segment breakdowns, trend analysis, recommendation): [later date]
>
> Which would be more useful given your timeline? I can always start with the quick version and follow up with the full one."

---

### Template 6: Follow-Up After Delivery

> "Hi [Name] — just checking in on the [topic] analysis I sent [day]. A couple questions:
>
> 1. Did it answer your question, or do you need me to dig further?
> 2. Anything that surprised you or didn't match your expectations?
> 3. Is this something you'll need again? (If so, I can explore automating it.)
>
> No rush — just want to make sure it was useful."

## 8. What to Avoid — Anti-Patterns

### Analysis Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| Starting to query before understanding the question | You'll answer the wrong thing and redo the work | Spend 5 minutes decoding and scoping first. Always. |
| Boiling the ocean | Exploring every possible angle instead of answering the specific question | Write a 5-line work plan. Time-box each step. |
| Delivering data without narrative | A spreadsheet full of numbers doesn't tell anyone what to do | Every deliverable needs a "bottom line" and a "so what" |
| Sending raw SQL output as the deliverable | Looks unprofessional, hard to read, no context | At minimum: formatted table, labeled columns, a header explaining what it shows |
| Not QA-ing before sending | One wrong number nukes your credibility | 5-minute QA ladder, every time, no exceptions |
| Over-building for a quick question | Spending 8 hours on something that needed 30 minutes | Match effort to stakes. Use the depth-vs-speed table. |
| Perfectionism before v1 | Polishing charts when the stakeholder just needs a directional answer | "Good enough to be trusted" is the bar for ad-hoc. Ship, then refine if asked. |

### Communication Anti-Patterns

| Bad Pattern | Why It's Bad | Better Alternative |
|-------------|-------------|-------------------|
| Answering the literal question without context | "Revenue was 2.1M" — ok, so what? Good? Bad? | Always add comparison: vs. target, vs. prior period, vs. expectation |
| Burying the lead | Starting with methodology instead of the answer | Bottom line first. Always. Methodology goes at the end or in an appendix. |
| No recommendation | "Here's the data, you decide" | Even a tentative recommendation shows judgment. "Based on this, I'd suggest X — but I'd want to validate with [further analysis]." |
| Not following up | You delivered and moved on | A 2-minute follow-up check builds the relationship and surfaces unmet needs |
| Over-caveating | So many disclaimers the stakeholder doesn't trust anything | 2–3 caveats max. State them clearly, then state your finding with confidence. |
| Sending at 11pm | Makes you look disorganized or burned out | Schedule send for 8am. Your analysis will land better when people are fresh. |

## 9. Worked Example: "What's Going On With Churn?"

This walks through a realistic ad-hoc request end-to-end, showing how the framework applies.

---

**The request (Monday 2pm, Slack DM from VP of Customer Success):**
> "Hey, can you look into churn? It feels like it's been ticking up. I might need to present something to leadership next week."

### Phase 1: Decode (10 minutes)

**Surface request:** "Look into churn"
**Your decoding questions (via Slack):**
> "Happy to dig into this. A few quick things:
> - When you say churn, do you mean logo churn (accounts lost) or dollar churn (revenue lost)?
> - What time horizon are you thinking? Last quarter? Last year?
> - What's the 'ticking up' based on — something you saw in a report, or a gut feeling from conversations?
> - If this goes to leadership, what decision would it inform?"

**VP's response:**
> "Dollar churn. I feel like the last 3-4 months have been rough. My team is mentioning more cancellations than usual. Leadership is going to want to know if we need to shift budget from acquisition to retention."

**Your decode:**
- Real question: "Is dollar churn trending up, and if so, where and why?"
- Decision: Budget allocation between acquisition and retention
- Audience: Leadership (needs to be polished and defensible)

### Phase 2: Scope (5 minutes)

**Your scope confirmation:**
> "Got it. Here's what I'll deliver:
>
> **Question:** Is dollar churn trending up over the last 6 months, and which segments are driving it?
> **Time period:** Last 6 months vs. prior 6 months (with 12-month trendline for context)
> **Segments:** Enterprise vs. SMB, by plan type and account tenure
> **Format:** 3-4 charts with summary bullets (deck-ready)
> **ETA:** Wednesday EOD
>
> Sound right?"

VP confirms.

### Phase 3: Explore (1.5 hours)

You pull the data. Initial findings:
- Overall dollar churn: up from 3.1% to 4.8% annualized over last 6 months
- Enterprise is driving it (5.2% → 7.1%), while SMB is flat (2.8% → 3.0%)
- Within enterprise, a specific cohort jumps out: accounts on the legacy pricing plan

You note the legacy plan finding but don't chase it yet — it's not in your work plan. You add it to "further investigation."

### Phase 4: Analyze (2 hours)

You break enterprise churn down by:
- Plan type: Legacy plan churns at 2.3x the rate of the new plan
- Account tenure: Accounts in months 12–18 have the highest churn spike
- Company size: Mid-market (50–200 employees) is the epicenter

Your SCR structure takes shape:
- **Situation:** Enterprise churn target is < 3% annualized
- **Complication:** Enterprise churn is 7.1%, driven by legacy-plan mid-market accounts in the 12–18 month tenure window
- **Resolution:** Recommend piloting a plan migration for the 20 highest-risk legacy accounts. If churn in this cohort drops to new-plan levels, enterprise churn would fall to ~4.2%.

### Phase 5: QA (30 minutes)

| Check | Result |
|-------|--------|
| Smell test | 7.1% enterprise churn feels high but plausible given VP's observations |
| Date boundaries | Confirmed: 6-month windows aligned to calendar months |
| Cross-reference | March churn figure matches Finance monthly report within 3K |
| NULLs | 8 accounts with NULL plan_type — investigated: all are in migration. Excluded from segment breakdown, included in totals. Noted in caveats. |
| Segments sum to total | Enterprise + SMB churn dollars = total. Confirmed. |

### Phase 6: Deliver (45 minutes)

You send an email (Template 4):

> **Subject: Churn Analysis — Enterprise churn up 2x, driven by legacy plan accounts**
>
> Hi [VP],
>
> **Bottom line:** Enterprise dollar churn has risen from 3.1% to 7.1% annualized over the last 6 months, driven almost entirely by mid-market accounts on the legacy pricing plan.
>
> **Key findings:**
> 1. Enterprise churn is 7.1% annualized (vs. 3.1% six months ago). SMB is flat at ~3%.
> 2. Accounts on the legacy plan churn at 2.3x the rate of accounts on the new plan.
> 3. The highest-risk window is months 12–18 of account tenure — this is when annual renewals hit and legacy pricing gaps become visible.
>
> **Recommendation:** Pilot a proactive plan migration for the top 20 at-risk legacy accounts. If this cohort's churn drops to new-plan levels, enterprise churn would fall to ~4.2%.
>
> **Caveats:** 8 accounts currently in migration excluded from segment breakdowns (included in totals). Mobile usage data not available before June 2025.
>
> Charts attached. Happy to walk through live before your leadership meeting."

### Phase 7: Follow up (Thursday)

> "Hi [VP] — just checking in. Did the churn analysis answer your question for the leadership meeting? Anything I should adjust or dig deeper on?"

VP responds: "This was exactly what I needed. Leadership approved the pilot migration. Can you help us track the results?"

**Outcome:** A vague Slack DM became a strategy decision in 4 days. Total effort: ~6 hours.

## 10. Quick-Reference Checklists

### When a Request Arrives

- [ ] Identify the surface request, real question, and underlying decision
- [ ] Ask clarifying questions before starting (use Template 1)
- [ ] Confirm: Can you state the decision this informs in one sentence?
- [ ] Assess depth level: quick pull, structured summary, full analysis, or deep dive
- [ ] Send scope confirmation (use Template 2)
- [ ] Get written acknowledgment before starting work

### During the Analysis

- [ ] Write a 5-line work plan with time boxes
- [ ] Pull core metric first — confirm data exists and makes sense
- [ ] Break down by likely drivers
- [ ] Test your leading hypothesis
- [ ] Apply the rabbit hole test to any tangent: does it change the answer?
- [ ] Track "further investigation" items separately — don't chase them now

### Before You Send

- [ ] Does the number pass the smell test?
- [ ] Date boundaries verified
- [ ] NULLs accounted for
- [ ] At least one cross-reference check against a known source
- [ ] Filters applied correctly
- [ ] Every number has comparison context (vs. what?)
- [ ] Assumptions and caveats stated
- [ ] Deliverable includes: bottom line, key findings, recommendation
- [ ] Would you bet 100 dollars this is right?

### After You Send

## 11. Final Takeaway

> **The one sentence to remember:** The best ad-hoc analysts don't start with data — they start with "what decision are you trying to make?"

---

**The script to internalize:**

*"Before I query: what's the real question, who's it for, and what will they do with the answer?*

*During the analysis: follow the work plan, resist rabbit holes, and time-box every step.*

*Before I send: QA the numbers, lead with the bottom line, and always include a recommendation."*

---

**Why this matters:** Ad-hoc work is where trust is built or broken. The analyst who consistently turns vague requests into clear, trusted answers becomes the person stakeholders rely on. The analyst who sends raw data without narrative, answers the wrong question, or lets ad-hocs drag on for weeks becomes the person stakeholders route around. The difference isn't talent — it's process.
